```python
# ================================================================
# ================================================================
# FILE WEAK
# rho = 0.20
#
# PEMBANGKITAN DATA SIMULASI SPASIAL
#
# DESAIN:
# Grid          : 10x10 s.d. 50x50
# Predictor     : X1, X2, X3
# Latent        : Z(s), TIDAK masuk model
#
# Pattern Z(s):
# 1. Hotspot  : pusat besar + pusat kecil
# 2. Koridor  : berkelok, tidak monoton
# 3. Gelombang: multikomponen
#
# g(X):
# sin(1.4X1)
# + 0.6(X2^2 - 1)
# + 0.6 tanh(X3)
#
# eta:
# 5 Z(s) + 0.8 g(X) + e
#
# SAR:
# Y = (I-rho W)^(-1) eta
#
# W:
# KNN5
# Distance Band <= 1.5
# IDW radius 2.5
#
# Error:
# Homogeneous
# Heterogeneous
#
# Untuk setiap grid:
# 3 pattern × 3 W × 2 error = 18 skenario
# ================================================================
# ================================================================


# ================================================================
# 0. LIBRARY
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import gc
import zipfile

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import matplotlib as mpl

import scipy.sparse as sp

from scipy.spatial import cKDTree
from scipy.sparse.linalg import spsolve
from scipy.interpolate import UnivariateSpline

import statsmodels.api as sm

from statsmodels.stats.diagnostic import (
    het_breuschpagan
)

from pathlib import Path


# ================================================================
# DISPLAY
# ================================================================

pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.width",
    3000
)

pd.set_option(
    "display.max_colwidth",
    None
)


# ================================================================
# 1. KONFIGURASI WEAK
# ================================================================

RHO_WEAK = 0.20

DELTA_SPATIAL = 5.0

LAMBDA_X = 0.80

ERROR_SD = 0.25

SPLINE_SMOOTH = 0.60


PATTERNS = [

    "Hotspot",

    "Koridor",

    "Gelombang"

]


W_METHODS = [

    "KNN5",

    "DISTANCE",

    "IDW"

]


ERROR_TYPES = [

    "Homogeneous",

    "Heterogeneous"

]


MODEL_PREDICTORS = [

    "X1",

    "X2",

    "X3"

]


# ================================================================
# FOLDER OUTPUT
# ================================================================

BASE_OUTPUT_WEAK = Path(

    "/content/"
    "SIMULASI_WEAK_RHO020"

)


BASE_OUTPUT_WEAK.mkdir(

    parents=True,

    exist_ok=True

)


# ================================================================
# 2. STANDARDIZE
# ================================================================

def standardize(v):

    v = np.asarray(
        v,
        dtype=float
    )

    sd = np.std(
        v,
        ddof=0
    )

    if sd <= 1e-12:

        return (
            v
            -
            np.mean(v)
        )

    return (

        (
            v
            -
            np.mean(v)
        )

        /
        sd

    )


# ================================================================
# 3. ROW STANDARDIZATION W
# ================================================================

def row_standardize(W):

    W = (

        W
        .tocsr()
        .astype(float)

    )


    row_sum = np.asarray(

        W.sum(
            axis=1
        )

    ).ravel()


    inverse = np.zeros_like(

        row_sum,

        dtype=float

    )


    good = (

        row_sum > 0

    )


    inverse[
        good
    ] = (

        1.0

        /

        row_sum[
            good
        ]

    )


    return (

        sp.diags(
            inverse
        )

        @

        W

    ).tocsr()


# ================================================================
# 4. GAUSSIAN HOTSPOT
# ================================================================

def gaussian_hotspot(
    x,
    y,
    cx,
    cy,
    sigma,
    amplitude=1.0
):

    return (

        amplitude

        *

        np.exp(

            -(

                (
                    x - cx
                ) ** 2

                +

                (
                    y - cy
                ) ** 2

            )

            /

            (
                2
                *
                sigma ** 2
            )

        )

    )


# ================================================================
# 5. SMOOTH SPLINE
# ================================================================

def smooth_spline_curve(
    x,
    y,
    smooth_fraction=0.60
):

    temp = pd.DataFrame({

        "x":
            np.asarray(x),

        "y":
            np.asarray(y)

    })


    temp[
        "x_round"
    ] = (

        temp[
            "x"
        ]

        .round(8)

    )


    temp = (

        temp

        .groupby(
            "x_round",
            as_index=False
        )

        .agg(

            x=(
                "x",
                "mean"
            ),

            y=(
                "y",
                "mean"
            )

        )

        .sort_values(
            "x"
        )

    )


    xs = (

        temp[
            "x"
        ]

        .to_numpy()

    )


    ys = (

        temp[
            "y"
        ]

        .to_numpy()

    )


    smoothing = (

        smooth_fraction

        *

        len(xs)

        *

        max(
            np.var(ys),
            1e-6
        )

    )


    spline = UnivariateSpline(

        xs,

        ys,

        k=3,

        s=smoothing

    )


    xx = np.linspace(

        xs.min(),

        xs.max(),

        400

    )


    yy = spline(
        xx
    )


    return (
        xx,
        yy
    )


# ================================================================
# 6. MORAN I
# ================================================================

def moran_I(
    y,
    W
):

    y = np.asarray(
        y,
        dtype=float
    )


    z = (

        y

        -

        np.mean(y)

    )


    denominator = float(

        z @ z

    )


    if denominator <= 0:

        return np.nan


    numerator = float(

        z

        @

        (
            W @ z
        )

    )


    S0 = float(

        W.sum()

    )


    return float(

        (
            len(y)
            /
            S0
        )

        *

        (
            numerator
            /
            denominator
        )

    )


# ================================================================
# ================================================================
# FUNGSI UTAMA
# SATU GRID = SEMUA PROSES + SEMUA VISUAL + VALIDASI
# ================================================================
# ================================================================

def run_weak_grid(
    grid_size
):


    # ============================================================
    # IDENTITAS
    # ============================================================

    rho = RHO_WEAK

    N = (

        grid_size

        *

        grid_size

    )


    OUTPUT_DIR = (

        BASE_OUTPUT_WEAK

        /

        f"GRID{grid_size}"

    )


    OUTPUT_DIR.mkdir(

        parents=True,

        exist_ok=True

    )


    print("\n")
    print("=" * 120)

    print(
        f"WEAK SPATIAL DEPENDENCE — rho={rho:.2f}"
    )

    print(
        f"GRID {grid_size} × {grid_size}"
    )

    print(
        f"N = {N}"
    )

    print("=" * 120)


    # ============================================================
    # A. GRID
    # ============================================================

    grid_row = np.repeat(

        np.arange(
            grid_size
        ),

        grid_size

    )


    grid_col = np.tile(

        np.arange(
            grid_size
        ),

        grid_size

    )


    x_raw = (
        grid_col
        .astype(float)
    )


    y_raw = (
        grid_row
        .astype(float)
    )


    coords_raw = np.column_stack([

        x_raw,

        y_raw

    ])


    x_norm = (

        x_raw

        /

        (
            grid_size - 1
        )

    )


    y_norm = (

        y_raw

        /

        (
            grid_size - 1
        )

    )


    GRID = pd.DataFrame({

        "Location_ID":
            np.arange(
                1,
                N + 1
            ),

        "Grid_Row":
            grid_row,

        "Grid_Col":
            grid_col,

        "X_norm":
            x_norm,

        "Y_norm":
            y_norm

    })


    print(
        "\n✅ A. GRID BERHASIL"
    )


    # ============================================================
    # B. HOTSPOT
    # ============================================================

    hotspot_big = gaussian_hotspot(

        x_norm,
        y_norm,

        cx=0.32,
        cy=0.56,

        sigma=0.17,

        amplitude=1.80

    )


    hotspot_small = gaussian_hotspot(

        x_norm,
        y_norm,

        cx=0.74,
        cy=0.76,

        sigma=0.085,

        amplitude=1.30

    )


    hotspot_irregular = (

        0.08

        *

        np.sin(
            3
            *
            np.pi
            *
            x_norm
        )

        *

        np.cos(
            2
            *
            np.pi
            *
            y_norm
        )

    )


    HOTSPOT_PARTS = {

        "besar":

            standardize(
                hotspot_big
            ),

        "kecil":

            standardize(
                hotspot_small
            ),

        "gabungan":

            standardize(

                hotspot_big

                +

                hotspot_small

                +

                hotspot_irregular

            )

    }


    # ============================================================
    # C. KORIDOR
    # ============================================================

    centerline = (

        0.20

        +

        0.54
        *
        x_norm

        +

        0.10
        *
        np.sin(

            2.6
            *
            np.pi
            *
            x_norm

            +
            0.40

        )

        -

        0.045
        *
        np.sin(

            5.4
            *
            np.pi
            *
            x_norm

        )

    )


    corridor_distance = np.abs(

        y_norm

        -

        centerline

    )


    corridor_width = (

        0.065

        +

        0.020

        *

        (
            1

            +

            np.sin(
                3
                *
                np.pi
                *
                x_norm
            )

        )

    )


    corridor_main = np.exp(

        -(

            corridor_distance ** 2

        )

        /

        (
            2
            *
            corridor_width ** 2
        )

    )


    taper_left = (

        1

        /

        (
            1

            +

            np.exp(

                -(

                    x_norm
                    -
                    0.10

                )

                /
                0.025

            )

        )

    )


    taper_right = (

        1

        /

        (
            1

            +

            np.exp(

                (

                    x_norm
                    -
                    0.90

                )

                /
                0.025

            )

        )

    )


    corridor_main = (

        corridor_main

        *

        taper_left

        *

        taper_right

    )


    corridor_bump1 = gaussian_hotspot(

        x_norm,
        y_norm,

        cx=0.28,
        cy=0.38,

        sigma=0.085,

        amplitude=0.45

    )


    corridor_bump2 = gaussian_hotspot(

        x_norm,
        y_norm,

        cx=0.72,
        cy=0.67,

        sigma=0.070,

        amplitude=0.35

    )


    corridor_local = (

        corridor_bump1

        +

        corridor_bump2

    )


    corridor_roughness = (

        0.08

        *

        np.sin(

            5
            *
            np.pi
            *
            x_norm

            +

            1.2
            *
            y_norm

        )

    )


    CORRIDOR_PARTS = {

        "utama":

            standardize(
                corridor_main
            ),

        "lokal":

            standardize(
                corridor_local
            ),

        "gabungan":

            standardize(

                1.15
                *
                corridor_main

                +

                corridor_local

                +

                corridor_roughness

            )

    }


    # ============================================================
    # D. GELOMBANG
    # ============================================================

    wave_large = (

        1.15

        *

        np.sin(

            2.5
            *
            np.pi
            *
            x_norm

            +
            0.30

        )

        +

        0.95

        *

        np.cos(

            2.1
            *
            np.pi
            *
            y_norm

            -
            0.20

        )

    )


    wave_secondary = (

        0.55

        *

        np.sin(

            4.0
            *
            np.pi
            *
            x_norm

            -

            1.7
            *
            np.pi
            *
            y_norm

        )

    )


    wave_diagonal = (

        0.40

        *

        np.sin(

            2
            *
            np.pi
            *
            (
                x_norm
                +
                y_norm
            )

            +
            0.50

        )

    )


    WAVE_PARTS = {

        "besar":

            standardize(
                wave_large
            ),

        "kecil":

            standardize(

                wave_secondary

                +

                wave_diagonal

            ),

        "gabungan":

            standardize(

                wave_large

                +

                wave_secondary

                +

                wave_diagonal

            )

    }


    Z_PATTERN = {

        "Hotspot":

            HOTSPOT_PARTS[
                "gabungan"
            ],

        "Koridor":

            CORRIDOR_PARTS[
                "gabungan"
            ],

        "Gelombang":

            WAVE_PARTS[
                "gabungan"
            ]

    }


    print(
        "✅ B. Z(s) SELESAI"
    )


    # ============================================================
    # VISUAL 1 — POLA Z(s)
    # ============================================================

    fig, axes = plt.subplots(

        3,
        3,

        figsize=(
            12,
            12
        )

    )


    visual_components = [

        (
            "Hotspot",

            HOTSPOT_PARTS[
                "besar"
            ],

            HOTSPOT_PARTS[
                "kecil"
            ],

            HOTSPOT_PARTS[
                "gabungan"
            ]

        ),

        (
            "Koridor",

            CORRIDOR_PARTS[
                "utama"
            ],

            CORRIDOR_PARTS[
                "lokal"
            ],

            CORRIDOR_PARTS[
                "gabungan"
            ]

        ),

        (
            "Gelombang",

            WAVE_PARTS[
                "besar"
            ],

            WAVE_PARTS[
                "kecil"
            ],

            WAVE_PARTS[
                "gabungan"
            ]

        )

    ]


    titles = [

        "Struktur Besar/Utama",

        "Struktur Kecil/Lokal",

        "Gabungan Z(s)"

    ]


    for r, (
        pattern,
        part1,
        part2,
        combined
    ) in enumerate(
        visual_components
    ):


        for c, values in enumerate(

            [
                part1,
                part2,
                combined
            ]

        ):


            matrix = values.reshape(

                grid_size,

                grid_size

            )


            im = axes[
                r,
                c
            ].imshow(

                matrix,

                origin="lower",

                cmap="viridis",

                interpolation="nearest"

            )


            if r == 0:

                axes[
                    r,
                    c
                ].set_title(

                    titles[c],

                    fontweight="bold"

                )


            if c == 0:

                axes[
                    r,
                    c
                ].set_ylabel(

                    pattern,

                    fontweight="bold"

                )


            axes[
                r,
                c
            ].set_xticks([])


            axes[
                r,
                c
            ].set_yticks([])


            plt.colorbar(

                im,

                ax=axes[
                    r,
                    c
                ],

                fraction=.046,

                pad=.04

            )


    fig.suptitle(

        (
            f"Latent Spatial Pattern Z(s)\n"
            f"Grid {grid_size}×{grid_size} | Weak rho=0.20"
        ),

        fontsize=16,

        fontweight="bold"

    )


    plt.tight_layout()


    FILE_VIS1 = (

        OUTPUT_DIR

        /

        "01_LATENT_PATTERN.png"

    )


    plt.savefig(

        FILE_VIS1,

        dpi=200,

        bbox_inches="tight"

    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # E. X1 X2 X3
    # ============================================================

    rng_x = np.random.default_rng(

        2026
        +
        grid_size

    )


    X_DATA = rng_x.normal(

        0,

        1,

        size=(
            N,
            3
        )

    )


    for j in range(3):

        X_DATA[
            :,
            j
        ] = standardize(

            X_DATA[
                :,
                j
            ]

        )


    X1 = X_DATA[
        :,
        0
    ]


    X2 = X_DATA[
        :,
        1
    ]


    X3 = X_DATA[
        :,
        2
    ]


    GX_RAW = (

        np.sin(

            1.4
            *
            X1

        )

        +

        0.60

        *

        (
            X2 ** 2

            -

            1

        )

        +

        0.60

        *

        np.tanh(
            X3
        )

    )


    GX = standardize(
        GX_RAW
    )


    print(
        "✅ C. X1, X2, X3 + g(X) SELESAI"
    )


    # ============================================================
    # VISUAL 2 — TRUE NONLINEARITY
    # ============================================================

    fig, axes = plt.subplots(

        1,
        3,

        figsize=(
            15,
            4.5
        )

    )


    X_DICT = {

        "X1":
            X1,

        "X2":
            X2,

        "X3":
            X3

    }


    for ax, xname in zip(

        axes,

        MODEL_PREDICTORS

    ):


        xx_line, yy_line = smooth_spline_curve(

            X_DICT[
                xname
            ],

            GX,

            smooth_fraction=0.25

        )


        ax.scatter(

            X_DICT[
                xname
            ],

            GX,

            alpha=.30,

            s=18

        )


        ax.plot(

            xx_line,

            yy_line,

            linewidth=3

        )


        ax.set_title(

            f"{xname} vs g(X)",

            fontweight="bold"

        )


        ax.set_xlabel(
            xname
        )


        ax.set_ylabel(
            "g(X)"
        )


        ax.grid(
            alpha=.20
        )


    fig.suptitle(

        (
            "Cek Nonlinearitas Sebenarnya pada DGP\n"
            f"Grid {grid_size}×{grid_size}"
        ),

        fontsize=15,

        fontweight="bold"

    )


    plt.tight_layout()


    FILE_VIS2 = (

        OUTPUT_DIR

        /

        "02_TRUE_NONLINEAR_SPLINE.png"

    )


    plt.savefig(

        FILE_VIS2,

        dpi=200,

        bbox_inches="tight"

    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # F. BUILD W
    # ============================================================

    tree = cKDTree(
        coords_raw
    )


    # ============================================================
    # KNN5
    # ============================================================

    distances_knn, indices_knn = tree.query(

        coords_raw,

        k=6

    )


    rows_knn = np.repeat(

        np.arange(N),

        5

    )


    cols_knn = (

        indices_knn[
            :,
            1:6
        ]

        .reshape(-1)

    )


    vals_knn = np.ones(

        len(
            rows_knn
        )

    )


    W_KNN5 = sp.csr_matrix(

        (
            vals_knn,

            (
                rows_knn,
                cols_knn
            )

        ),

        shape=(
            N,
            N
        )

    )


    W_KNN5 = row_standardize(
        W_KNN5
    )


    # ============================================================
    # DISTANCE BAND
    # ============================================================

    pairs_distance = np.array(

        list(

            tree.query_pairs(
                r=1.5
            )

        ),

        dtype=int

    )


    if len(
        pairs_distance
    ) > 0:


        rows_d = np.concatenate([

            pairs_distance[
                :,
                0
            ],

            pairs_distance[
                :,
                1
            ]

        ])


        cols_d = np.concatenate([

            pairs_distance[
                :,
                1
            ],

            pairs_distance[
                :,
                0
            ]

        ])


        vals_d = np.ones(

            len(
                rows_d
            )

        )


    else:


        rows_d = np.array(
            [],
            dtype=int
        )


        cols_d = np.array(
            [],
            dtype=int
        )


        vals_d = np.array(
            [],
            dtype=float
        )


    W_DISTANCE = sp.csr_matrix(

        (
            vals_d,

            (
                rows_d,
                cols_d
            )

        ),

        shape=(
            N,
            N
        )

    )


    W_DISTANCE = row_standardize(
        W_DISTANCE
    )


    # ============================================================
    # IDW
    # ============================================================

    pairs_idw = np.array(

        list(

            tree.query_pairs(
                r=2.5
            )

        ),

        dtype=int

    )


    if len(
        pairs_idw
    ) > 0:


        i1 = pairs_idw[
            :,
            0
        ]


        i2 = pairs_idw[
            :,
            1
        ]


        dist_idw = np.linalg.norm(

            coords_raw[
                i1
            ]

            -

            coords_raw[
                i2
            ],

            axis=1

        )


        weight_idw = (

            1.0

            /

            dist_idw

        )


        rows_i = np.concatenate([

            i1,

            i2

        ])


        cols_i = np.concatenate([

            i2,

            i1

        ])


        vals_i = np.concatenate([

            weight_idw,

            weight_idw

        ])


    else:


        rows_i = np.array(
            [],
            dtype=int
        )


        cols_i = np.array(
            [],
            dtype=int
        )


        vals_i = np.array(
            [],
            dtype=float
        )


    W_IDW = sp.csr_matrix(

        (
            vals_i,

            (
                rows_i,
                cols_i
            )

        ),

        shape=(
            N,
            N
        )

    )


    W_IDW = row_standardize(
        W_IDW
    )


    W_STORE = {

        "KNN5":
            W_KNN5,

        "DISTANCE":
            W_DISTANCE,

        "IDW":
            W_IDW

    }


    W_INFO_ROWS = []


    for method in W_METHODS:


        W = W_STORE[
            method
        ]


        counts = np.diff(

            W.indptr

        )


        W_INFO_ROWS.append({

            "Grid":
                f"{grid_size}x{grid_size}",

            "W_Method":
                method,

            "Min_Neighbor":
                int(
                    counts.min()
                ),

            "Mean_Neighbor":
                float(
                    counts.mean()
                ),

            "Max_Neighbor":
                int(
                    counts.max()
                )

        })


    W_INFO = pd.DataFrame(
        W_INFO_ROWS
    )


    print(
        "✅ D. W SELESAI"
    )


    display(

        W_INFO.round(2)

    )


    # ============================================================
    # VISUAL 3 — W
    # ============================================================

    target_index = np.argmin(

        (
            x_norm
            -
            0.5
        ) ** 2

        +

        (
            y_norm
            -
            0.5
        ) ** 2

    )


    fig, axes = plt.subplots(

        1,
        3,

        figsize=(
            15,
            5
        )

    )


    for ax, method in zip(

        axes,

        W_METHODS

    ):


        W = W_STORE[
            method
        ].tocsr()


        start = W.indptr[
            target_index
        ]


        end = W.indptr[
            target_index + 1
        ]


        neighbors = W.indices[
            start:end
        ]


        ax.scatter(

            x_raw,

            y_raw,

            s=12,

            alpha=.20

        )


        ax.scatter(

            x_raw[
                neighbors
            ],

            y_raw[
                neighbors
            ],

            s=70,

            label="Tetangga"

        )


        ax.scatter(

            x_raw[
                target_index
            ],

            y_raw[
                target_index
            ],

            marker="*",

            s=220,

            label="Lokasi pusat"

        )


        for j in neighbors:


            ax.plot(

                [
                    x_raw[
                        target_index
                    ],

                    x_raw[j]
                ],

                [
                    y_raw[
                        target_index
                    ],

                    y_raw[j]
                ],

                linewidth=.8,

                alpha=.50

            )


        title_w = {

            "KNN5":
                "KNN-5",

            "DISTANCE":
                "Distance Band",

            "IDW":
                "IDW"

        }[
            method
        ]


        ax.set_title(

            (
                f"{title_w}\n"
                f"k={len(neighbors)}"
            ),

            fontweight="bold"

        )


        ax.set_aspect(
            "equal"
        )


        ax.legend(
            fontsize=7
        )


    fig.suptitle(

        (
            "Struktur Ketetanggaan W\n"
            f"Grid {grid_size}×{grid_size}"
        ),

        fontsize=15,

        fontweight="bold"

    )


    plt.tight_layout()


    FILE_VIS3 = (

        OUTPUT_DIR

        /

        "03_STRUKTUR_W.png"

    )


    plt.savefig(

        FILE_VIS3,

        dpi=200,

        bbox_inches="tight"

    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # G. ERROR
    # ============================================================

    rng_error = np.random.default_rng(

        2027
        +
        grid_size

    )


    BASE_ERROR = rng_error.normal(

        0,

        1,

        N

    )


    def generate_error(
        error_type
    ):


        if (
            error_type
            ==
            "Homogeneous"
        ):


            variance = np.ones(
                N
            )


            error = (

                ERROR_SD

                *

                BASE_ERROR

            )


        else:


            variance = np.where(

                x_norm
                <=
                0.5,

                0.5,

                1.5

            )


            error = (

                ERROR_SD

                *

                np.sqrt(
                    variance
                )

                *

                BASE_ERROR

            )


        return (

            error,

            variance

        )


    # ============================================================
    # H. GENERATE 18 SKENARIO
    # ============================================================

    ALL_DATA = []


    for pattern in PATTERNS:


        Z_SPATIAL = Z_PATTERN[
            pattern
        ]


        SPATIAL_SIGNAL = (

            DELTA_SPATIAL

            *

            Z_SPATIAL

        )


        for error_type in ERROR_TYPES:


            ERROR, ERROR_VARIANCE = (

                generate_error(
                    error_type
                )

            )


            ETA = (

                SPATIAL_SIGNAL

                +

                LAMBDA_X
                *
                GX

                +

                ERROR

            )


            for w_method in W_METHODS:


                W = W_STORE[
                    w_method
                ]


                A = (

                    sp.eye(
                        N,
                        format="csr"
                    )

                    -

                    rho
                    *
                    W

                )


                Y = spsolve(

                    A,

                    ETA

                )


                Y = np.asarray(

                    Y,

                    dtype=float

                )


                MORAN_Y = moran_I(

                    Y,

                    W

                )


                CORR_ZY = float(

                    np.corrcoef(

                        Z_SPATIAL,

                        Y

                    )[0, 1]

                )


                scenario_id = (

                    f"G{grid_size}_"

                    f"{pattern.upper()}_"

                    f"{w_method}_"

                    f"{'HOMO' if error_type == 'Homogeneous' else 'HETERO'}_"

                    "RHO20"

                )


                temp = GRID.copy()


                temp[
                    "Scenario_ID"
                ] = scenario_id


                temp[
                    "Grid"
                ] = (

                    f"{grid_size}x{grid_size}"

                )


                temp[
                    "N"
                ] = N


                temp[
                    "Strength"
                ] = "Weak"


                temp[
                    "Rho"
                ] = rho


                temp[
                    "Pattern"
                ] = pattern


                temp[
                    "W_Method"
                ] = w_method


                temp[
                    "Error_Structure"
                ] = error_type


                temp[
                    "Z_spatial_latent"
                ] = Z_SPATIAL


                temp[
                    "X1"
                ] = X1


                temp[
                    "X2"
                ] = X2


                temp[
                    "X3"
                ] = X3


                temp[
                    "gX_nonlinear"
                ] = GX


                temp[
                    "Spatial_Signal"
                ] = SPATIAL_SIGNAL


                temp[
                    "Error"
                ] = ERROR


                temp[
                    "Error_Variance"
                ] = ERROR_VARIANCE


                temp[
                    "Eta"
                ] = ETA


                temp[
                    "Y"
                ] = Y


                temp[
                    "Moran_I_Y"
                ] = MORAN_Y


                temp[
                    "Corr_Z_Y"
                ] = CORR_ZY


                ALL_DATA.append(
                    temp
                )


    DATA = pd.concat(

        ALL_DATA,

        ignore_index=True

    )


    assert (

        DATA[
            "Scenario_ID"
        ].nunique()

        ==
        18

    )


    print(
        "✅ E. 18 SKENARIO Y SELESAI"
    )


    # ============================================================
    # VISUAL 4 — 18 Y
    # ============================================================

    rows_visual = [

        (
            p,
            e
        )

        for p in PATTERNS

        for e in ERROR_TYPES

    ]


    vmin = float(

        DATA[
            "Y"
        ].min()

    )


    vmax = float(

        DATA[
            "Y"
        ].max()

    )


    norm = mpl.colors.Normalize(

        vmin=vmin,

        vmax=vmax

    )


    fig, axes = plt.subplots(

        6,
        3,

        figsize=(
            14,
            24
        )

    )


    last_im = None


    for r, (
        pattern,
        error_type
    ) in enumerate(
        rows_visual
    ):


        for c, w_method in enumerate(
            W_METHODS
        ):


            ax = axes[
                r,
                c
            ]


            d = (

                DATA[

                    (
                        DATA[
                            "Pattern"
                        ]
                        ==
                        pattern
                    )

                    &

                    (
                        DATA[
                            "Error_Structure"
                        ]
                        ==
                        error_type
                    )

                    &

                    (
                        DATA[
                            "W_Method"
                        ]
                        ==
                        w_method
                    )

                ]

                .sort_values(
                    "Location_ID"
                )

            )


            Y_MATRIX = (

                d[
                    "Y"
                ]

                .to_numpy()

                .reshape(

                    grid_size,

                    grid_size

                )

            )


            last_im = ax.imshow(

                Y_MATRIX,

                origin="lower",

                cmap="viridis",

                norm=norm,

                interpolation="nearest"

            )


            if r == 0:


                ax.set_title(

                    {

                        "KNN5":
                            "KNN-5",

                        "DISTANCE":
                            "Distance Band",

                        "IDW":
                            "IDW"

                    }[
                        w_method
                    ],

                    fontweight="bold"

                )


            if c == 0:


                error_short = (

                    "Homo"

                    if (
                        error_type
                        ==
                        "Homogeneous"
                    )

                    else
                    "Hetero"

                )


                ax.set_ylabel(

                    (
                        f"{pattern}\n"
                        f"{error_short}"
                    ),

                    fontweight="bold"

                )


            if (
                error_type
                ==
                "Heterogeneous"
            ):


                ax.axvline(

                    grid_size/2
                    -
                    .5,

                    linestyle="--",

                    linewidth=1.5

                )


            W = W_STORE[
                w_method
            ]


            kmean = np.diff(

                W.indptr

            ).mean()


            ax.text(

                .5,

                -.08,

                (
                    f"ρ={rho:.2f}"
                    f" | k̄={kmean:.2f}"
                    "\n"
                    f"Moran I={d['Moran_I_Y'].iloc[0]:.3f}"
                    f" | Corr(Z,Y)={d['Corr_Z_Y'].iloc[0]:.3f}"
                ),

                transform=ax.transAxes,

                ha="center",

                fontsize=7.5

            )


            ax.set_xticks([])

            ax.set_yticks([])


    fig.colorbar(

        last_im,

        ax=axes,

        fraction=.015,

        pad=.02,

        label="Nilai Y"

    )


    fig.suptitle(

        (
            f"18 Skenario Y — Grid {grid_size}×{grid_size}\n"
            "WEAK Spatial Dependence | rho=0.20"
        ),

        fontsize=17,

        fontweight="bold",

        y=.995

    )


    FILE_VIS4 = (

        OUTPUT_DIR

        /

        "04_Y_18_SKENARIO.png"

    )


    plt.savefig(

        FILE_VIS4,

        dpi=180,

        bbox_inches="tight"

    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # VISUAL 5 — Z VS Y
    # ============================================================

    fig, axes = plt.subplots(

        3,
        2,

        figsize=(
            8,
            12
        )

    )


    for r, pattern in enumerate(
        PATTERNS
    ):


        d = (

            DATA[

                (
                    DATA[
                        "Pattern"
                    ]
                    ==
                    pattern
                )

                &

                (
                    DATA[
                        "W_Method"
                    ]
                    ==
                    "KNN5"
                )

                &

                (
                    DATA[
                        "Error_Structure"
                    ]
                    ==
                    "Homogeneous"
                )

            ]

            .sort_values(
                "Location_ID"
            )

        )


        ZMAT = (

            d[
                "Z_spatial_latent"
            ]

            .to_numpy()

            .reshape(

                grid_size,

                grid_size

            )

        )


        YMAT = (

            d[
                "Y"
            ]

            .to_numpy()

            .reshape(

                grid_size,

                grid_size

            )

        )


        im1 = axes[
            r,
            0
        ].imshow(

            ZMAT,

            origin="lower",

            cmap="viridis",

            interpolation="nearest"

        )


        axes[
            r,
            0
        ].set_title(

            (
                f"{pattern}\n"
                "Z(s) dibangkitkan"
            ),

            fontweight="bold"

        )


        plt.colorbar(

            im1,

            ax=axes[
                r,
                0
            ],

            fraction=.046,

            pad=.04

        )


        im2 = axes[
            r,
            1
        ].imshow(

            YMAT,

            origin="lower",

            cmap="viridis",

            interpolation="nearest"

        )


        axes[
            r,
            1
        ].set_title(

            (
                f"{pattern}\n"
                f"Y | Corr={d['Corr_Z_Y'].iloc[0]:.3f}"
            ),

            fontweight="bold"

        )


        plt.colorbar(

            im2,

            ax=axes[
                r,
                1
            ],

            fraction=.046,

            pad=.04

        )


        for c in range(2):


            axes[
                r,
                c
            ].set_xticks([])


            axes[
                r,
                c
            ].set_yticks([])


    fig.suptitle(

        (
            "Kesesuaian Pola Z(s) → Y\n"
            f"Grid {grid_size}×{grid_size}"
        ),

        fontsize=15,

        fontweight="bold"

    )


    plt.tight_layout()


    FILE_VIS5 = (

        OUTPUT_DIR

        /

        "05_Z_VS_Y.png"

    )


    plt.savefig(

        FILE_VIS5,

        dpi=200,

        bbox_inches="tight"

    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # VISUAL 6 — SMOOTH SPLINE X VS Y
    # ============================================================

    fig, axes = plt.subplots(

        3,
        3,

        figsize=(
            15,
            13
        )

    )


    for r, pattern in enumerate(
        PATTERNS
    ):


        d = (

            DATA[

                (
                    DATA[
                        "Pattern"
                    ]
                    ==
                    pattern
                )

                &

                (
                    DATA[
                        "W_Method"
                    ]
                    ==
                    "KNN5"
                )

                &

                (
                    DATA[
                        "Error_Structure"
                    ]
                    ==
                    "Homogeneous"
                )

            ]

            .sort_values(
                "Location_ID"
            )

        )


        for c, xname in enumerate(
            MODEL_PREDICTORS
        ):


            x_values = (

                d[
                    xname
                ]

                .to_numpy()

            )


            y_values = (

                d[
                    "Y"
                ]

                .to_numpy()

            )


            xx_line, yy_line = smooth_spline_curve(

                x_values,

                y_values,

                smooth_fraction=
                    SPLINE_SMOOTH

            )


            axes[
                r,
                c
            ].scatter(

                x_values,

                y_values,

                alpha=.25,

                s=18

            )


            axes[
                r,
                c
            ].plot(

                xx_line,

                yy_line,

                linewidth=3

            )


            if r == 0:


                axes[
                    r,
                    c
                ].set_title(

                    xname,

                    fontweight="bold"

                )


            if c == 0:


                axes[
                    r,
                    c
                ].set_ylabel(

                    (
                        f"{pattern}\n"
                        "Y"
                    ),

                    fontweight="bold"

                )


            axes[
                r,
                c
            ].set_xlabel(
                xname
            )


            axes[
                r,
                c
            ].grid(
                alpha=.20
            )


    fig.suptitle(

        (
            "Smooth Spline Prediktor terhadap Y\n"
            f"Grid {grid_size}×{grid_size} | KNN5 | Homo | rho=.20"
        ),

        fontsize=15,

        fontweight="bold"

    )


    plt.tight_layout()


    FILE_VIS6 = (

        OUTPUT_DIR

        /

        "06_SMOOTH_SPLINE_X_Y.png"

    )


    plt.savefig(

        FILE_VIS6,

        dpi=200,

        bbox_inches="tight"

    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # ============================================================
    # I. VALIDASI BREUSCH-PAGAN
    # ============================================================
    # ============================================================

    VALIDATION_ROWS = []


    for scenario_id in (

        DATA[
            "Scenario_ID"
        ]

        .drop_duplicates()

    ):


        d = (

            DATA[

                DATA[
                    "Scenario_ID"
                ]
                ==
                scenario_id

            ]

            .sort_values(
                "Location_ID"
            )

            .reset_index(
                drop=True
            )

        )


        # ========================================================
        # residual true DGP = error
        # ========================================================

        true_mean = (

            d[
                "Spatial_Signal"
            ].to_numpy()

            +

            LAMBDA_X

            *

            d[
                "gX_nonlinear"
            ].to_numpy()

        )


        residual_dgp = (

            d[
                "Eta"
            ].to_numpy()

            -

            true_mean

        )


        residual_match = np.allclose(

            residual_dgp,

            d[
                "Error"
            ].to_numpy(),

            atol=1e-10

        )


        # ========================================================
        # ZONA
        # ========================================================

        zone2 = (

            d[
                "X_norm"
            ].to_numpy()

            >
            0.5

        ).astype(int)


        X_BP = sm.add_constant(
            zone2
        )


        # ========================================================
        # BREUSCH PAGAN
        # ========================================================

        (
            BP_LM,
            BP_LM_pvalue,
            BP_F,
            BP_F_pvalue

        ) = het_breuschpagan(

            residual_dgp,

            X_BP

        )


        error_type = (

            d[
                "Error_Structure"
            ].iloc[0]

        )


        # ========================================================
        # EXPECTED
        # ========================================================

        if (
            error_type
            ==
            "Homogeneous"
        ):


            BP_Expected = (
                "Homoskedastik"
            )


            BP_Status = (

                "PASS"

                if (

                    BP_LM_pvalue > 0.05

                    and

                    BP_F_pvalue > 0.05

                )

                else "CHECK"

            )


        else:


            BP_Expected = (
                "Heteroskedastik"
            )


            BP_Status = (

                "PASS"

                if (

                    BP_LM_pvalue < 0.05

                    and

                    BP_F_pvalue < 0.05

                )

                else "CHECK"

            )


        # ========================================================
        # ACTUAL DECISION
        # ========================================================

        BP_Decision = (

            "Heteroskedastik"

            if BP_LM_pvalue < 0.05

            else

            "Homoskedastik"

        )


        # ========================================================
        # VARIANS ZONA
        # ========================================================

        Var_Zone1 = np.var(

            residual_dgp[
                zone2 == 0
            ],

            ddof=1

        )


        Var_Zone2 = np.var(

            residual_dgp[
                zone2 == 1
            ],

            ddof=1

        )


        Var_Ratio = (

            Var_Zone2

            /

            Var_Zone1

        )


        if (
            error_type
            ==
            "Homogeneous"
        ):


            Variance_Status = (

                "PASS"

                if (

                    0.50

                    <=

                    Var_Ratio

                    <=

                    2.00

                )

                else "CHECK"

            )


        else:


            Variance_Status = (

                "PASS"

                if Var_Ratio > 1.50

                else "CHECK"

            )


        moran_value = float(

            d[
                "Moran_I_Y"
            ].iloc[0]

        )


        corr_value = float(

            d[
                "Corr_Z_Y"
            ].iloc[0]

        )


        Moran_Status = (

            "PASS"

            if moran_value > 0

            else "CHECK"

        )


        Pattern_Status = (

            "PASS"

            if corr_value >= 0.75

            else "CHECK"

        )


        VALIDATION_ROWS.append({

            "Pattern":
                d[
                    "Pattern"
                ].iloc[0],

            "W_Method":
                d[
                    "W_Method"
                ].iloc[0],

            "Error_Structure":
                error_type,

            "Rho":
                rho,

            "BP_LM":
                BP_LM,

            "BP_LM_pvalue":
                BP_LM_pvalue,

            "BP_F":
                BP_F,

            "BP_F_pvalue":
                BP_F_pvalue,

            "BP_Expected":
                BP_Expected,

            "BP_Decision":
                BP_Decision,

            "BP_Status":
                BP_Status,

            "Var_Zone1":
                Var_Zone1,

            "Var_Zone2":
                Var_Zone2,

            "Var_Ratio":
                Var_Ratio,

            "Variance_Status":
                Variance_Status,

            "Moran_I_Y":
                moran_value,

            "Moran_Status":
                Moran_Status,

            "Corr_Z_Y":
                corr_value,

            "Pattern_Status":
                Pattern_Status,

            "Residual_Match":
                residual_match

        })


    VALIDATION = pd.DataFrame(

        VALIDATION_ROWS

    )


    # ============================================================
    # ============================================================
    # CEK BREUSCH-PAGAN
    #
    # FORMAT MIRIP SCREENSHOT CEK TREND MORAN
    # ============================================================
    # ============================================================

    BP_CHECK = (

        VALIDATION[

            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "BP_LM",
                "BP_LM_pvalue",
                "BP_F",
                "BP_F_pvalue",
                "BP_Expected",
                "BP_Decision",
                "BP_Status"
            ]

        ]

        .sort_values(

            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]

        )

        .reset_index(
            drop=True
        )

    )


    print("\n")
    print("=" * 130)

    print(
        f"CEK BREUSCH-PAGAN — GRID {grid_size}×{grid_size} — WEAK rho=0.20:"
    )

    print("=" * 130)


    print(

        BP_CHECK

        .round({

            "BP_LM":
                4,

            "BP_LM_pvalue":
                6,

            "BP_F":
                4,

            "BP_F_pvalue":
                6

        })

        .to_string(
            index=False
        )

    )


    # ============================================================
    # VARIANS ZONA
    # ============================================================

    VAR_CHECK = (

        VALIDATION[

            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Var_Zone1",
                "Var_Zone2",
                "Var_Ratio",
                "Variance_Status"
            ]

        ]

        .sort_values(

            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]

        )

        .reset_index(
            drop=True
        )

    )


    print("\n")
    print(
        "CEK VARIANS ZONA:"
    )


    print(

        VAR_CHECK

        .round({

            "Var_Zone1":
                6,

            "Var_Zone2":
                6,

            "Var_Ratio":
                4

        })

        .to_string(
            index=False
        )

    )


    # ============================================================
    # ============================================================
    # CEK MORAN WEAK
    # ============================================================
    # ============================================================

    MORAN_CHECK = (

        VALIDATION[

            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Rho",
                "Moran_I_Y",
                "Moran_Status"
            ]

        ]

        .sort_values(

            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]

        )

        .reset_index(
            drop=True
        )

    )


    print("\n")
    print("=" * 100)

    print(
        f"CEK MORAN — GRID {grid_size}×{grid_size} — WEAK:"
    )

    print("=" * 100)


    print(

        MORAN_CHECK

        .round({

            "Rho":
                2,

            "Moran_I_Y":
                4

        })

        .to_string(
            index=False
        )

    )


    # ============================================================
    # ============================================================
    # KESESUAIAN POLA Z(s) → Y
    # ============================================================
    # ============================================================

    PATTERN_CHECK = (

        VALIDATION[

            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Corr_Z_Y",
                "Pattern_Status"
            ]

        ]

        .sort_values(

            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]

        )

        .reset_index(
            drop=True
        )

    )


    print("\n")
    print("=" * 100)

    print(
        f"KESESUAIAN POLA Z(s) → Y — GRID {grid_size}×{grid_size}:"
    )

    print("=" * 100)


    print(

        PATTERN_CHECK

        .round({

            "Corr_Z_Y":
                4

        })

        .to_string(
            index=False
        )

    )


    # ============================================================
    # FINAL STATUS
    # ============================================================

    BP_ALL_PASS = (

        VALIDATION[
            "BP_Status"
        ]

        .eq(
            "PASS"
        )

        .all()

    )


    VAR_ALL_PASS = (

        VALIDATION[
            "Variance_Status"
        ]

        .eq(
            "PASS"
        )

        .all()

    )


    MORAN_ALL_PASS = (

        VALIDATION[
            "Moran_Status"
        ]

        .eq(
            "PASS"
        )

        .all()

    )


    PATTERN_ALL_PASS = (

        VALIDATION[
            "Pattern_Status"
        ]

        .eq(
            "PASS"
        )

        .all()

    )


    # ============================================================
    # SUMMARY GRID
    # ============================================================

    homo = (

        VALIDATION[

            VALIDATION[
                "Error_Structure"
            ]
            ==
            "Homogeneous"

        ]

        .iloc[0]

    )


    hetero = (

        VALIDATION[

            VALIDATION[
                "Error_Structure"
            ]
            ==
            "Heterogeneous"

        ]

        .iloc[0]

    )


    SUMMARY_GRID = pd.DataFrame({

        "Grid": [
            f"{grid_size}x{grid_size}"
        ],

        "N": [
            N
        ],

        "Rho": [
            rho
        ],

        "Scenario": [
            18
        ],

        "BP_Homo_LM": [
            homo[
                "BP_LM"
            ]
        ],

        "BP_Homo_p": [
            homo[
                "BP_LM_pvalue"
            ]
        ],

        "BP_Homo_Decision": [
            homo[
                "BP_Decision"
            ]
        ],

        "BP_Hetero_LM": [
            hetero[
                "BP_LM"
            ]
        ],

        "BP_Hetero_p": [
            hetero[
                "BP_LM_pvalue"
            ]
        ],

        "BP_Hetero_Decision": [
            hetero[
                "BP_Decision"
            ]
        ],

        "Homo_Var_Ratio": [
            homo[
                "Var_Ratio"
            ]
        ],

        "Hetero_Var_Ratio": [
            hetero[
                "Var_Ratio"
            ]
        ],

        "Mean_Moran_I": [
            VALIDATION[
                "Moran_I_Y"
            ].mean()
        ],

        "Min_Moran_I": [
            VALIDATION[
                "Moran_I_Y"
            ].min()
        ],

        "Max_Moran_I": [
            VALIDATION[
                "Moran_I_Y"
            ].max()
        ],

        "Mean_Corr_Z_Y": [
            VALIDATION[
                "Corr_Z_Y"
            ].mean()
        ],

        "Min_Corr_Z_Y": [
            VALIDATION[
                "Corr_Z_Y"
            ].min()
        ],

        "Max_Corr_Z_Y": [
            VALIDATION[
                "Corr_Z_Y"
            ].max()
        ],

        "BP_All_PASS": [
            BP_ALL_PASS
        ],

        "Variance_All_PASS": [
            VAR_ALL_PASS
        ],

        "Moran_All_PASS": [
            MORAN_ALL_PASS
        ],

        "Pattern_All_PASS": [
            PATTERN_ALL_PASS
        ]

    })


    print("\n")
    print("=" * 140)

    print(
        f"SUMMARY FINAL GRID {grid_size}×{grid_size}"
    )

    print("=" * 140)


    print(

        SUMMARY_GRID

        .round({

            "Rho":
                2,

            "BP_Homo_LM":
                4,

            "BP_Homo_p":
                6,

            "BP_Hetero_LM":
                4,

            "BP_Hetero_p":
                6,

            "Homo_Var_Ratio":
                4,

            "Hetero_Var_Ratio":
                4,

            "Mean_Moran_I":
                4,

            "Min_Moran_I":
                4,

            "Max_Moran_I":
                4,

            "Mean_Corr_Z_Y":
                4,

            "Min_Corr_Z_Y":
                4,

            "Max_Corr_Z_Y":
                4

        })

        .to_string(
            index=False
        )

    )


    # ============================================================
    # SAVE
    # ============================================================

    DATA_FILE = (

        OUTPUT_DIR

        /

        f"DATA_WEAK_GRID{grid_size}_RHO020.csv"

    )


    DATA.to_csv(

        DATA_FILE,

        index=False

    )


    VALIDATION_FILE = (

        OUTPUT_DIR

        /

        f"VALIDASI_WEAK_GRID{grid_size}_RHO020.csv"

    )


    VALIDATION.to_csv(

        VALIDATION_FILE,

        index=False

    )


    BP_FILE = (

        OUTPUT_DIR

        /

        f"BREUSCH_PAGAN_GRID{grid_size}.csv"

    )


    BP_CHECK.to_csv(

        BP_FILE,

        index=False

    )


    SUMMARY_FILE = (

        OUTPUT_DIR

        /

        f"SUMMARY_WEAK_GRID{grid_size}.csv"

    )


    SUMMARY_GRID.to_csv(

        SUMMARY_FILE,

        index=False

    )


    W_INFO.to_csv(

        OUTPUT_DIR

        /

        f"INFO_W_GRID{grid_size}.csv",

        index=False

    )


    for method in W_METHODS:


        sp.save_npz(

            OUTPUT_DIR

            /

            f"W_GRID{grid_size}_{method}.npz",

            W_STORE[
                method
            ]

        )


    print("\n")
    print("=" * 100)

    print(
        f"🎉 GRID {grid_size}×{grid_size} WEAK SELESAI"
    )

    print("=" * 100)


    print(
        "Breusch-Pagan :",
        "PASS ✅"
        if BP_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Varians Zona  :",
        "PASS ✅"
        if VAR_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Moran's I     :",
        "PASS ✅"
        if MORAN_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Pola Z(s)→Y   :",
        "PASS ✅"
        if PATTERN_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "\n📁",
        OUTPUT_DIR
    )


    gc.collect()


    return {

        "grid":
            grid_size,

        "data":
            DATA,

        "validation":
            VALIDATION,

        "bp":
            BP_CHECK,

        "moran":
            MORAN_CHECK,

        "pattern_check":
            PATTERN_CHECK,

        "summary":
            SUMMARY_GRID,

        "output":
            OUTPUT_DIR

    }
```

```python
# ================================================================
# ================================================================
# MODERATE SPATIAL DEPENDENCE
# rho = 0.50
#
# CELL 1
# GRID 10×10
#
# SEKALIGUS MENDEFINISIKAN MESIN SIMULASI
# UNTUK GRID 10 - 50
#
# Setiap grid menghasilkan:
#
# 1. Z(s) Hotspot / Koridor / Gelombang
# 2. Visual seluruh bentuk Z(s)
# 3. X1, X2, X3
# 4. True nonlinear smooth spline
# 5. KNN5 / Distance / IDW
# 6. Visual W
# 7. Homo / Hetero
# 8. 18 skenario Y
# 9. Visual 18 Y
# 10. Z(s) vs Y
# 11. Smooth spline X vs Y
# 12. Breusch-Pagan per skenario
# 13. Cek varians zona
# 14. Moran's I
# 15. Corr Z(s),Y
# 16. Summary final
# 17. Save semua file
# ================================================================
# ================================================================


# ================================================================
# 0. LIBRARY
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import gc
import zipfile

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import matplotlib as mpl

import scipy.sparse as sp

from scipy.spatial import cKDTree
from scipy.sparse.linalg import spsolve
from scipy.interpolate import UnivariateSpline

import statsmodels.api as sm

from statsmodels.stats.diagnostic import (
    het_breuschpagan
)

from pathlib import Path


# ================================================================
# DISPLAY
# ================================================================

pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.width",
    3000
)

pd.set_option(
    "display.max_colwidth",
    None
)


# ================================================================
# 1. KONFIGURASI MODERATE
# ================================================================

RHO_MODERATE = 0.50

DELTA_SPATIAL = 5.0

LAMBDA_X = 0.80

ERROR_SD = 0.25

SPLINE_SMOOTH = 0.60


PATTERNS = [
    "Hotspot",
    "Koridor",
    "Gelombang"
]


W_METHODS = [
    "KNN5",
    "DISTANCE",
    "IDW"
]


ERROR_TYPES = [
    "Homogeneous",
    "Heterogeneous"
]


MODEL_PREDICTORS = [
    "X1",
    "X2",
    "X3"
]


# ================================================================
# OUTPUT
# ================================================================

BASE_OUTPUT_MODERATE = Path(
    "/content/SIMULASI_MODERATE_RHO050"
)


BASE_OUTPUT_MODERATE.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# STANDARDIZE
# ================================================================

def standardize(v):

    v = np.asarray(
        v,
        dtype=float
    )

    sd = np.std(
        v,
        ddof=0
    )

    if sd <= 1e-12:

        return (
            v
            -
            np.mean(v)
        )

    return (
        (v - np.mean(v))
        /
        sd
    )


# ================================================================
# ROW STANDARDIZE W
# ================================================================

def row_standardize(W):

    W = (
        W
        .tocsr()
        .astype(float)
    )

    row_sum = np.asarray(
        W.sum(axis=1)
    ).ravel()

    inv = np.zeros_like(
        row_sum,
        dtype=float
    )

    valid = (
        row_sum > 0
    )

    inv[valid] = (
        1.0
        /
        row_sum[valid]
    )

    return (
        sp.diags(inv)
        @
        W
    ).tocsr()


# ================================================================
# GAUSSIAN HOTSPOT
# ================================================================

def gaussian_hotspot(
    x,
    y,
    cx,
    cy,
    sigma,
    amplitude=1.0
):

    return (
        amplitude
        *
        np.exp(
            -(
                (x-cx)**2
                +
                (y-cy)**2
            )
            /
            (
                2
                *
                sigma**2
            )
        )
    )


# ================================================================
# SMOOTH SPLINE
# ================================================================

def smooth_spline_curve(
    x,
    y,
    smooth_fraction=0.60
):

    temp = pd.DataFrame({
        "x": np.asarray(x),
        "y": np.asarray(y)
    })

    temp["x_round"] = (
        temp["x"]
        .round(8)
    )

    temp = (
        temp

        .groupby(
            "x_round",
            as_index=False
        )

        .agg(
            x=("x", "mean"),
            y=("y", "mean")
        )

        .sort_values("x")
    )

    xs = (
        temp["x"]
        .to_numpy()
    )

    ys = (
        temp["y"]
        .to_numpy()
    )

    smoothing = (
        smooth_fraction
        *
        len(xs)
        *
        max(
            np.var(ys),
            1e-6
        )
    )

    spline = UnivariateSpline(
        xs,
        ys,
        k=3,
        s=smoothing
    )

    xx = np.linspace(
        xs.min(),
        xs.max(),
        400
    )

    yy = spline(xx)

    return xx, yy


# ================================================================
# MORAN'S I
# ================================================================

def moran_I(
    y,
    W
):

    y = np.asarray(
        y,
        dtype=float
    )

    z = (
        y
        -
        np.mean(y)
    )

    denominator = float(
        z @ z
    )

    if denominator <= 0:

        return np.nan

    numerator = float(
        z
        @
        (W @ z)
    )

    S0 = float(
        W.sum()
    )

    return float(
        (len(y)/S0)
        *
        (numerator/denominator)
    )


# ================================================================
# ================================================================
# FUNGSI UTAMA MODERATE
# ================================================================
# ================================================================

def run_moderate_grid(
    grid_size
):

    rho = RHO_MODERATE

    N = (
        grid_size
        *
        grid_size
    )


    OUTPUT_DIR = (
        BASE_OUTPUT_MODERATE
        /
        f"GRID{grid_size}"
    )


    OUTPUT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )


    print("\n")
    print("=" * 120)

    print(
        "MODERATE SPATIAL DEPENDENCE"
    )

    print(
        f"rho = {rho:.2f}"
    )

    print(
        f"GRID {grid_size} × {grid_size}"
    )

    print(
        f"N = {N}"
    )

    print("=" * 120)


    # ============================================================
    # A. GRID
    # ============================================================

    grid_row = np.repeat(
        np.arange(grid_size),
        grid_size
    )


    grid_col = np.tile(
        np.arange(grid_size),
        grid_size
    )


    x_raw = (
        grid_col
        .astype(float)
    )


    y_raw = (
        grid_row
        .astype(float)
    )


    coords_raw = np.column_stack([
        x_raw,
        y_raw
    ])


    x_norm = (
        x_raw
        /
        (grid_size - 1)
    )


    y_norm = (
        y_raw
        /
        (grid_size - 1)
    )


    GRID = pd.DataFrame({

        "Location_ID":
            np.arange(
                1,
                N+1
            ),

        "Grid_Row":
            grid_row,

        "Grid_Col":
            grid_col,

        "X_norm":
            x_norm,

        "Y_norm":
            y_norm

    })


    # ============================================================
    # B. HOTSPOT
    # ============================================================

    hotspot_big = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.32,
        cy=0.56,
        sigma=0.17,
        amplitude=1.80
    )


    hotspot_small = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.74,
        cy=0.76,
        sigma=0.085,
        amplitude=1.30
    )


    hotspot_irregular = (
        0.08
        *
        np.sin(
            3*np.pi*x_norm
        )
        *
        np.cos(
            2*np.pi*y_norm
        )
    )


    HOTSPOT_PARTS = {

        "besar":
            standardize(
                hotspot_big
            ),

        "kecil":
            standardize(
                hotspot_small
            ),

        "gabungan":
            standardize(
                hotspot_big
                +
                hotspot_small
                +
                hotspot_irregular
            )
    }


    # ============================================================
    # C. KORIDOR
    # ============================================================

    centerline = (
        0.20

        +

        0.54
        *
        x_norm

        +

        0.10
        *
        np.sin(
            2.6*np.pi*x_norm
            +
            0.40
        )

        -

        0.045
        *
        np.sin(
            5.4*np.pi*x_norm
        )
    )


    corridor_distance = np.abs(
        y_norm
        -
        centerline
    )


    corridor_width = (
        0.065

        +

        0.020
        *
        (
            1
            +
            np.sin(
                3*np.pi*x_norm
            )
        )
    )


    corridor_main = np.exp(
        -(corridor_distance**2)
        /
        (
            2
            *
            corridor_width**2
        )
    )


    taper_left = (
        1
        /
        (
            1
            +
            np.exp(
                -(x_norm-0.10)
                /
                0.025
            )
        )
    )


    taper_right = (
        1
        /
        (
            1
            +
            np.exp(
                (x_norm-0.90)
                /
                0.025
            )
        )
    )


    corridor_main = (
        corridor_main
        *
        taper_left
        *
        taper_right
    )


    corridor_bump1 = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.28,
        cy=0.38,
        sigma=0.085,
        amplitude=0.45
    )


    corridor_bump2 = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.72,
        cy=0.67,
        sigma=0.070,
        amplitude=0.35
    )


    corridor_local = (
        corridor_bump1
        +
        corridor_bump2
    )


    corridor_roughness = (
        0.08
        *
        np.sin(
            5*np.pi*x_norm
            +
            1.2*y_norm
        )
    )


    CORRIDOR_PARTS = {

        "utama":
            standardize(
                corridor_main
            ),

        "lokal":
            standardize(
                corridor_local
            ),

        "gabungan":
            standardize(
                1.15*corridor_main
                +
                corridor_local
                +
                corridor_roughness
            )
    }


    # ============================================================
    # D. GELOMBANG
    # ============================================================

    wave_large = (
        1.15
        *
        np.sin(
            2.5*np.pi*x_norm
            +
            0.30
        )

        +

        0.95
        *
        np.cos(
            2.1*np.pi*y_norm
            -
            0.20
        )
    )


    wave_secondary = (
        0.55
        *
        np.sin(
            4.0*np.pi*x_norm
            -
            1.7*np.pi*y_norm
        )
    )


    wave_diagonal = (
        0.40
        *
        np.sin(
            2*np.pi
            *
            (
                x_norm
                +
                y_norm
            )
            +
            0.50
        )
    )


    WAVE_PARTS = {

        "besar":
            standardize(
                wave_large
            ),

        "kecil":
            standardize(
                wave_secondary
                +
                wave_diagonal
            ),

        "gabungan":
            standardize(
                wave_large
                +
                wave_secondary
                +
                wave_diagonal
            )
    }


    Z_PATTERN = {

        "Hotspot":
            HOTSPOT_PARTS[
                "gabungan"
            ],

        "Koridor":
            CORRIDOR_PARTS[
                "gabungan"
            ],

        "Gelombang":
            WAVE_PARTS[
                "gabungan"
            ]
    }


    # ============================================================
    # VISUAL 1 — Z(s)
    # ============================================================

    fig, axes = plt.subplots(
        3,
        3,
        figsize=(12,12)
    )


    VISUAL_COMPONENTS = [

        (
            "Hotspot",
            HOTSPOT_PARTS["besar"],
            HOTSPOT_PARTS["kecil"],
            HOTSPOT_PARTS["gabungan"]
        ),

        (
            "Koridor",
            CORRIDOR_PARTS["utama"],
            CORRIDOR_PARTS["lokal"],
            CORRIDOR_PARTS["gabungan"]
        ),

        (
            "Gelombang",
            WAVE_PARTS["besar"],
            WAVE_PARTS["kecil"],
            WAVE_PARTS["gabungan"]
        )

    ]


    titles = [
        "Struktur Besar/Utama",
        "Struktur Kecil/Lokal",
        "Gabungan Z(s)"
    ]


    for r, (
        pattern,
        p1,
        p2,
        combined
    ) in enumerate(
        VISUAL_COMPONENTS
    ):


        for c, values in enumerate(
            [
                p1,
                p2,
                combined
            ]
        ):


            mat = values.reshape(
                grid_size,
                grid_size
            )


            im = axes[
                r,
                c
            ].imshow(
                mat,
                origin="lower",
                cmap="viridis",
                interpolation="nearest"
            )


            if r == 0:

                axes[
                    r,
                    c
                ].set_title(
                    titles[c],
                    fontweight="bold"
                )


            if c == 0:

                axes[
                    r,
                    c
                ].set_ylabel(
                    pattern,
                    fontweight="bold"
                )


            axes[r,c].set_xticks([])

            axes[r,c].set_yticks([])


            plt.colorbar(
                im,
                ax=axes[r,c],
                fraction=.046,
                pad=.04
            )


    fig.suptitle(
        (
            f"Latent Spatial Pattern Z(s)\n"
            f"Grid {grid_size}×{grid_size} | Moderate rho=0.50"
        ),
        fontsize=16,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS1 = (
        OUTPUT_DIR
        /
        "01_LATENT_PATTERN.png"
    )


    plt.savefig(
        FILE_VIS1,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # E. X1 X2 X3
    # ============================================================

    # Seed SAMA seperti Weak untuk ukuran grid yang sama.
    # Jadi perbandingan rho menjadi lebih fair.

    rng_x = np.random.default_rng(
        2026
        +
        grid_size
    )


    X_DATA = rng_x.normal(
        0,
        1,
        size=(
            N,
            3
        )
    )


    for j in range(3):

        X_DATA[:,j] = standardize(
            X_DATA[:,j]
        )


    X1 = X_DATA[:,0]

    X2 = X_DATA[:,1]

    X3 = X_DATA[:,2]


    GX_RAW = (
        np.sin(
            1.4*X1
        )

        +

        0.60
        *
        (
            X2**2
            -
            1
        )

        +

        0.60
        *
        np.tanh(
            X3
        )
    )


    GX = standardize(
        GX_RAW
    )


    # ============================================================
    # VISUAL 2 — TRUE NONLINEARITY
    # ============================================================

    X_DICT = {
        "X1": X1,
        "X2": X2,
        "X3": X3
    }


    fig, axes = plt.subplots(
        1,
        3,
        figsize=(15,4.5)
    )


    for ax, xname in zip(
        axes,
        MODEL_PREDICTORS
    ):


        xx, yy = smooth_spline_curve(
            X_DICT[xname],
            GX,
            smooth_fraction=.25
        )


        ax.scatter(
            X_DICT[xname],
            GX,
            alpha=.30,
            s=18
        )


        ax.plot(
            xx,
            yy,
            linewidth=3
        )


        ax.set_title(
            f"{xname} vs g(X)",
            fontweight="bold"
        )


        ax.set_xlabel(xname)

        ax.set_ylabel("g(X)")

        ax.grid(alpha=.20)


    fig.suptitle(
        (
            "Cek Nonlinearitas Sebenarnya pada DGP\n"
            f"Grid {grid_size}×{grid_size}"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS2 = (
        OUTPUT_DIR
        /
        "02_TRUE_NONLINEAR_SPLINE.png"
    )


    plt.savefig(
        FILE_VIS2,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # F. W
    # ============================================================

    tree = cKDTree(
        coords_raw
    )


    # ---------------- KNN5 ----------------

    distances_knn, indices_knn = tree.query(
        coords_raw,
        k=6
    )


    rows_knn = np.repeat(
        np.arange(N),
        5
    )


    cols_knn = (
        indices_knn[:,1:6]
        .reshape(-1)
    )


    vals_knn = np.ones(
        len(rows_knn)
    )


    W_KNN5 = sp.csr_matrix(
        (
            vals_knn,
            (
                rows_knn,
                cols_knn
            )
        ),
        shape=(N,N)
    )


    W_KNN5 = row_standardize(
        W_KNN5
    )


    # ---------------- DISTANCE ----------------

    pairs_distance = np.array(
        list(
            tree.query_pairs(
                r=1.5
            )
        ),
        dtype=int
    )


    if len(pairs_distance) > 0:

        rows_d = np.concatenate([
            pairs_distance[:,0],
            pairs_distance[:,1]
        ])

        cols_d = np.concatenate([
            pairs_distance[:,1],
            pairs_distance[:,0]
        ])

        vals_d = np.ones(
            len(rows_d)
        )

    else:

        rows_d = np.array(
            [],
            dtype=int
        )

        cols_d = np.array(
            [],
            dtype=int
        )

        vals_d = np.array(
            [],
            dtype=float
        )


    W_DISTANCE = sp.csr_matrix(
        (
            vals_d,
            (
                rows_d,
                cols_d
            )
        ),
        shape=(N,N)
    )


    W_DISTANCE = row_standardize(
        W_DISTANCE
    )


    # ---------------- IDW ----------------

    pairs_idw = np.array(
        list(
            tree.query_pairs(
                r=2.5
            )
        ),
        dtype=int
    )


    if len(pairs_idw) > 0:

        i1 = pairs_idw[:,0]

        i2 = pairs_idw[:,1]


        dist_idw = np.linalg.norm(
            coords_raw[i1]
            -
            coords_raw[i2],
            axis=1
        )


        weight_idw = (
            1.0
            /
            dist_idw
        )


        rows_i = np.concatenate([
            i1,
            i2
        ])


        cols_i = np.concatenate([
            i2,
            i1
        ])


        vals_i = np.concatenate([
            weight_idw,
            weight_idw
        ])

    else:

        rows_i = np.array(
            [],
            dtype=int
        )

        cols_i = np.array(
            [],
            dtype=int
        )

        vals_i = np.array(
            [],
            dtype=float
        )


    W_IDW = sp.csr_matrix(
        (
            vals_i,
            (
                rows_i,
                cols_i
            )
        ),
        shape=(N,N)
    )


    W_IDW = row_standardize(
        W_IDW
    )


    W_STORE = {

        "KNN5":
            W_KNN5,

        "DISTANCE":
            W_DISTANCE,

        "IDW":
            W_IDW

    }


    W_INFO_ROWS = []


    for method in W_METHODS:


        counts = np.diff(
            W_STORE[
                method
            ].indptr
        )


        W_INFO_ROWS.append({

            "Grid":
                f"{grid_size}x{grid_size}",

            "W_Method":
                method,

            "Min_Neighbor":
                int(
                    counts.min()
                ),

            "Mean_Neighbor":
                float(
                    counts.mean()
                ),

            "Max_Neighbor":
                int(
                    counts.max()
                )

        })


    W_INFO = pd.DataFrame(
        W_INFO_ROWS
    )


    print("\nINFORMASI W:")

    print(
        W_INFO
        .round(2)
        .to_string(
            index=False
        )
    )


    # ============================================================
    # VISUAL 3 — W
    # ============================================================

    target_index = np.argmin(
        (
            x_norm-.5
        )**2
        +
        (
            y_norm-.5
        )**2
    )


    fig, axes = plt.subplots(
        1,
        3,
        figsize=(15,5)
    )


    for ax, method in zip(
        axes,
        W_METHODS
    ):


        W = W_STORE[
            method
        ].tocsr()


        start = W.indptr[
            target_index
        ]

        end = W.indptr[
            target_index+1
        ]


        neighbors = W.indices[
            start:end
        ]


        ax.scatter(
            x_raw,
            y_raw,
            s=12,
            alpha=.20
        )


        ax.scatter(
            x_raw[neighbors],
            y_raw[neighbors],
            s=70,
            label="Tetangga"
        )


        ax.scatter(
            x_raw[target_index],
            y_raw[target_index],
            marker="*",
            s=220,
            label="Lokasi pusat"
        )


        for j in neighbors:

            ax.plot(
                [
                    x_raw[target_index],
                    x_raw[j]
                ],
                [
                    y_raw[target_index],
                    y_raw[j]
                ],
                linewidth=.8,
                alpha=.50
            )


        ax.set_title(
            (
                f"{method}\n"
                f"k={len(neighbors)}"
            ),
            fontweight="bold"
        )


        ax.set_aspect(
            "equal"
        )


        ax.legend(
            fontsize=7
        )


    fig.suptitle(
        (
            "Struktur Ketetanggaan W\n"
            f"Grid {grid_size}×{grid_size}"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS3 = (
        OUTPUT_DIR
        /
        "03_STRUKTUR_W.png"
    )


    plt.savefig(
        FILE_VIS3,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # G. ERROR
    # ============================================================

    # Seed juga SAMA dengan Weak
    # agar perubahan antar-rho fair.

    rng_error = np.random.default_rng(
        2027
        +
        grid_size
    )


    BASE_ERROR = rng_error.normal(
        0,
        1,
        N
    )


    def generate_error(
        error_type
    ):


        if (
            error_type
            ==
            "Homogeneous"
        ):

            variance = np.ones(
                N
            )

            error = (
                ERROR_SD
                *
                BASE_ERROR
            )

        else:

            variance = np.where(
                x_norm <= .5,
                .5,
                1.5
            )

            error = (
                ERROR_SD
                *
                np.sqrt(
                    variance
                )
                *
                BASE_ERROR
            )


        return (
            error,
            variance
        )


    # ============================================================
    # H. GENERATE 18 SKENARIO
    # ============================================================

    ALL_DATA = []


    for pattern in PATTERNS:


        Z_SPATIAL = Z_PATTERN[
            pattern
        ]


        SPATIAL_SIGNAL = (
            DELTA_SPATIAL
            *
            Z_SPATIAL
        )


        for error_type in ERROR_TYPES:


            ERROR, ERROR_VARIANCE = (
                generate_error(
                    error_type
                )
            )


            ETA = (
                SPATIAL_SIGNAL
                +
                LAMBDA_X
                *
                GX
                +
                ERROR
            )


            for w_method in W_METHODS:


                W = W_STORE[
                    w_method
                ]


                A = (
                    sp.eye(
                        N,
                        format="csr"
                    )

                    -

                    rho
                    *
                    W
                )


                Y = spsolve(
                    A,
                    ETA
                )


                Y = np.asarray(
                    Y,
                    dtype=float
                )


                moran_y = moran_I(
                    Y,
                    W
                )


                corr_zy = float(
                    np.corrcoef(
                        Z_SPATIAL,
                        Y
                    )[0,1]
                )


                scenario_id = (
                    f"G{grid_size}_"
                    f"{pattern.upper()}_"
                    f"{w_method}_"
                    f"{'HOMO' if error_type == 'Homogeneous' else 'HETERO'}_"
                    f"RHO50"
                )


                temp = GRID.copy()


                temp[
                    "Scenario_ID"
                ] = scenario_id


                temp[
                    "Grid"
                ] = (
                    f"{grid_size}x{grid_size}"
                )


                temp[
                    "N"
                ] = N


                temp[
                    "Strength"
                ] = "Moderate"


                temp[
                    "Rho"
                ] = rho


                temp[
                    "Pattern"
                ] = pattern


                temp[
                    "W_Method"
                ] = w_method


                temp[
                    "Error_Structure"
                ] = error_type


                temp[
                    "Z_spatial_latent"
                ] = Z_SPATIAL


                temp["X1"] = X1

                temp["X2"] = X2

                temp["X3"] = X3


                temp[
                    "gX_nonlinear"
                ] = GX


                temp[
                    "Spatial_Signal"
                ] = SPATIAL_SIGNAL


                temp[
                    "Error"
                ] = ERROR


                temp[
                    "Error_Variance"
                ] = ERROR_VARIANCE


                temp[
                    "Eta"
                ] = ETA


                temp[
                    "Y"
                ] = Y


                temp[
                    "Moran_I_Y"
                ] = moran_y


                temp[
                    "Corr_Z_Y"
                ] = corr_zy


                ALL_DATA.append(
                    temp
                )


    DATA = pd.concat(
        ALL_DATA,
        ignore_index=True
    )


    assert (
        DATA[
            "Scenario_ID"
        ].nunique()
        ==
        18
    )


    assert (
        len(DATA)
        ==
        18*N
    )


    # ============================================================
    # VISUAL 4 — 18 Y
    # ============================================================

    rows_plot = [
        (
            p,
            e
        )
        for p in PATTERNS
        for e in ERROR_TYPES
    ]


    norm = mpl.colors.Normalize(
        vmin=float(
            DATA["Y"].min()
        ),
        vmax=float(
            DATA["Y"].max()
        )
    )


    fig, axes = plt.subplots(
        6,
        3,
        figsize=(14,24)
    )


    last_im = None


    for r, (
        pattern,
        error_type
    ) in enumerate(
        rows_plot
    ):


        for c, w_method in enumerate(
            W_METHODS
        ):


            d = (
                DATA[
                    (
                        DATA[
                            "Pattern"
                        ]
                        ==
                        pattern
                    )

                    &

                    (
                        DATA[
                            "Error_Structure"
                        ]
                        ==
                        error_type
                    )

                    &

                    (
                        DATA[
                            "W_Method"
                        ]
                        ==
                        w_method
                    )
                ]

                .sort_values(
                    "Location_ID"
                )
            )


            YMAT = (
                d["Y"]
                .to_numpy()
                .reshape(
                    grid_size,
                    grid_size
                )
            )


            ax = axes[
                r,
                c
            ]


            last_im = ax.imshow(
                YMAT,
                origin="lower",
                cmap="viridis",
                norm=norm,
                interpolation="nearest"
            )


            if r == 0:

                ax.set_title(
                    w_method,
                    fontweight="bold"
                )


            if c == 0:

                err = (
                    "Homo"
                    if error_type
                    ==
                    "Homogeneous"
                    else
                    "Hetero"
                )


                ax.set_ylabel(
                    (
                        f"{pattern}\n"
                        f"{err}"
                    ),
                    fontweight="bold"
                )


            if (
                error_type
                ==
                "Heterogeneous"
            ):

                ax.axvline(
                    grid_size/2-.5,
                    linestyle="--",
                    linewidth=1.5
                )


            counts = np.diff(
                W_STORE[
                    w_method
                ].indptr
            )


            ax.text(
                .5,
                -.08,
                (
                    f"ρ=.50 | "
                    f"k̄={counts.mean():.2f}"
                    "\n"
                    f"Moran I={d['Moran_I_Y'].iloc[0]:.3f}"
                    f" | Corr(Z,Y)={d['Corr_Z_Y'].iloc[0]:.3f}"
                ),
                transform=ax.transAxes,
                ha="center",
                fontsize=7.5
            )


            ax.set_xticks([])

            ax.set_yticks([])


    fig.colorbar(
        last_im,
        ax=axes,
        fraction=.015,
        pad=.02,
        label="Nilai Y"
    )


    fig.suptitle(
        (
            f"18 Skenario Y — Grid {grid_size}×{grid_size}\n"
            "MODERATE Spatial Dependence | rho=0.50"
        ),
        fontsize=17,
        fontweight="bold",
        y=.995
    )


    FILE_VIS4 = (
        OUTPUT_DIR
        /
        "04_Y_18_SKENARIO.png"
    )


    plt.savefig(
        FILE_VIS4,
        dpi=180,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # VISUAL 5 — Z(s) VS Y
    # ============================================================

    fig, axes = plt.subplots(
        3,
        2,
        figsize=(8,12)
    )


    for r, pattern in enumerate(
        PATTERNS
    ):


        d = (
            DATA[
                (
                    DATA[
                        "Pattern"
                    ]
                    ==
                    pattern
                )

                &

                (
                    DATA[
                        "W_Method"
                    ]
                    ==
                    "KNN5"
                )

                &

                (
                    DATA[
                        "Error_Structure"
                    ]
                    ==
                    "Homogeneous"
                )
            ]

            .sort_values(
                "Location_ID"
            )
        )


        ZMAT = (
            d[
                "Z_spatial_latent"
            ]
            .to_numpy()
            .reshape(
                grid_size,
                grid_size
            )
        )


        YMAT = (
            d["Y"]
            .to_numpy()
            .reshape(
                grid_size,
                grid_size
            )
        )


        im1 = axes[
            r,
            0
        ].imshow(
            ZMAT,
            origin="lower",
            cmap="viridis"
        )


        axes[
            r,
            0
        ].set_title(
            (
                f"{pattern}\n"
                "Z(s)"
            ),
            fontweight="bold"
        )


        plt.colorbar(
            im1,
            ax=axes[r,0],
            fraction=.046,
            pad=.04
        )


        im2 = axes[
            r,
            1
        ].imshow(
            YMAT,
            origin="lower",
            cmap="viridis"
        )


        axes[
            r,
            1
        ].set_title(
            (
                f"{pattern}\n"
                f"Y | Corr="
                f"{d['Corr_Z_Y'].iloc[0]:.3f}"
            ),
            fontweight="bold"
        )


        plt.colorbar(
            im2,
            ax=axes[r,1],
            fraction=.046,
            pad=.04
        )


        for c in range(2):

            axes[
                r,
                c
            ].set_xticks([])

            axes[
                r,
                c
            ].set_yticks([])


    fig.suptitle(
        (
            "Kesesuaian Pola Z(s) → Y\n"
            f"Grid {grid_size}×{grid_size} | Moderate"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS5 = (
        OUTPUT_DIR
        /
        "05_Z_VS_Y.png"
    )


    plt.savefig(
        FILE_VIS5,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # VISUAL 6 — SMOOTH SPLINE X vs Y
    # ============================================================

    fig, axes = plt.subplots(
        3,
        3,
        figsize=(15,13)
    )


    for r, pattern in enumerate(
        PATTERNS
    ):


        d = (
            DATA[
                (
                    DATA[
                        "Pattern"
                    ]
                    ==
                    pattern
                )

                &

                (
                    DATA[
                        "W_Method"
                    ]
                    ==
                    "KNN5"
                )

                &

                (
                    DATA[
                        "Error_Structure"
                    ]
                    ==
                    "Homogeneous"
                )
            ]

            .sort_values(
                "Location_ID"
            )
        )


        for c, xname in enumerate(
            MODEL_PREDICTORS
        ):


            xv = (
                d[xname]
                .to_numpy()
            )


            yv = (
                d["Y"]
                .to_numpy()
            )


            xx, yy = smooth_spline_curve(
                xv,
                yv,
                smooth_fraction=
                    SPLINE_SMOOTH
            )


            axes[
                r,
                c
            ].scatter(
                xv,
                yv,
                alpha=.25,
                s=18
            )


            axes[
                r,
                c
            ].plot(
                xx,
                yy,
                linewidth=3
            )


            if r == 0:

                axes[
                    r,
                    c
                ].set_title(
                    xname,
                    fontweight="bold"
                )


            if c == 0:

                axes[
                    r,
                    c
                ].set_ylabel(
                    (
                        f"{pattern}\n"
                        "Y"
                    ),
                    fontweight="bold"
                )


            axes[
                r,
                c
            ].set_xlabel(
                xname
            )


            axes[
                r,
                c
            ].grid(
                alpha=.20
            )


    fig.suptitle(
        (
            "Smooth Spline Prediktor terhadap Y\n"
            f"Grid {grid_size}×{grid_size} | KNN5 | Homo | rho=.50"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS6 = (
        OUTPUT_DIR
        /
        "06_SMOOTH_SPLINE_X_Y.png"
    )


    plt.savefig(
        FILE_VIS6,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # ============================================================
    # I. VALIDASI
    # ============================================================
    # ============================================================

    VALIDATION_ROWS = []


    for scenario_id in (
        DATA[
            "Scenario_ID"
        ].drop_duplicates()
    ):


        d = (
            DATA[
                DATA[
                    "Scenario_ID"
                ]
                ==
                scenario_id
            ]

            .sort_values(
                "Location_ID"
            )

            .reset_index(
                drop=True
            )
        )


        true_mean = (
            d[
                "Spatial_Signal"
            ].to_numpy()

            +

            LAMBDA_X
            *
            d[
                "gX_nonlinear"
            ].to_numpy()
        )


        residual = (
            d["Eta"]
            .to_numpy()

            -

            true_mean
        )


        zone2 = (
            d[
                "X_norm"
            ].to_numpy()
            >
            .5
        ).astype(int)


        X_BP = sm.add_constant(
            zone2
        )


        (
            BP_LM,
            BP_LM_p,
            BP_F,
            BP_F_p
        ) = het_breuschpagan(
            residual,
            X_BP
        )


        error_type = (
            d[
                "Error_Structure"
            ].iloc[0]
        )


        if (
            error_type
            ==
            "Homogeneous"
        ):

            BP_EXPECTED = (
                "Homoskedastik"
            )

            BP_STATUS = (
                "PASS"
                if (
                    BP_LM_p > .05
                    and
                    BP_F_p > .05
                )
                else
                "CHECK"
            )

        else:

            BP_EXPECTED = (
                "Heteroskedastik"
            )

            BP_STATUS = (
                "PASS"
                if (
                    BP_LM_p < .05
                    and
                    BP_F_p < .05
                )
                else
                "CHECK"
            )


        BP_DECISION = (
            "Heteroskedastik"
            if BP_LM_p < .05
            else
            "Homoskedastik"
        )


        var1 = np.var(
            residual[
                zone2 == 0
            ],
            ddof=1
        )


        var2 = np.var(
            residual[
                zone2 == 1
            ],
            ddof=1
        )


        var_ratio = (
            var2
            /
            var1
        )


        if (
            error_type
            ==
            "Homogeneous"
        ):

            VAR_STATUS = (
                "PASS"
                if (
                    .50
                    <=
                    var_ratio
                    <=
                    2.00
                )
                else
                "CHECK"
            )

        else:

            VAR_STATUS = (
                "PASS"
                if var_ratio > 1.50
                else
                "CHECK"
            )


        moran_value = float(
            d[
                "Moran_I_Y"
            ].iloc[0]
        )


        corr_value = float(
            d[
                "Corr_Z_Y"
            ].iloc[0]
        )


        VALIDATION_ROWS.append({

            "Pattern":
                d[
                    "Pattern"
                ].iloc[0],

            "W_Method":
                d[
                    "W_Method"
                ].iloc[0],

            "Error_Structure":
                error_type,

            "Rho":
                rho,

            "BP_LM":
                BP_LM,

            "BP_LM_pvalue":
                BP_LM_p,

            "BP_F":
                BP_F,

            "BP_F_pvalue":
                BP_F_p,

            "BP_Expected":
                BP_EXPECTED,

            "BP_Decision":
                BP_DECISION,

            "BP_Status":
                BP_STATUS,

            "Var_Zone1":
                var1,

            "Var_Zone2":
                var2,

            "Var_Ratio":
                var_ratio,

            "Variance_Status":
                VAR_STATUS,

            "Moran_I_Y":
                moran_value,

            "Moran_Status":
                (
                    "PASS"
                    if moran_value > 0
                    else
                    "CHECK"
                ),

            "Corr_Z_Y":
                corr_value,

            "Pattern_Status":
                (
                    "PASS"
                    if corr_value >= .75
                    else
                    "CHECK"
                )

        })


    VALIDATION = pd.DataFrame(
        VALIDATION_ROWS
    )


    # ============================================================
    # BP TABLE
    # ============================================================

    BP_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "BP_LM",
                "BP_LM_pvalue",
                "BP_F",
                "BP_F_pvalue",
                "BP_Expected",
                "BP_Decision",
                "BP_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print("\n")
    print("=" * 130)

    print(
        f"CEK BREUSCH-PAGAN — GRID {grid_size}×{grid_size} — MODERATE rho=0.50:"
    )

    print("=" * 130)


    print(
        BP_CHECK

        .round({
            "BP_LM":4,
            "BP_LM_pvalue":6,
            "BP_F":4,
            "BP_F_pvalue":6
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # VARIANCE TABLE
    # ============================================================

    VAR_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Var_Zone1",
                "Var_Zone2",
                "Var_Ratio",
                "Variance_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print("\nCEK VARIANS ZONA:")


    print(
        VAR_CHECK

        .round({
            "Var_Zone1":6,
            "Var_Zone2":6,
            "Var_Ratio":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # MORAN TABLE
    # ============================================================

    MORAN_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Rho",
                "Moran_I_Y",
                "Moran_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print("\n")
    print("=" * 100)

    print(
        f"CEK MORAN — GRID {grid_size}×{grid_size} — MODERATE:"
    )

    print("=" * 100)


    print(
        MORAN_CHECK

        .round({
            "Rho":2,
            "Moran_I_Y":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # PATTERN TABLE
    # ============================================================

    PATTERN_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Corr_Z_Y",
                "Pattern_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print("\n")
    print("=" * 100)

    print(
        f"KESESUAIAN POLA Z(s) → Y — GRID {grid_size}×{grid_size}:"
    )

    print("=" * 100)


    print(
        PATTERN_CHECK

        .round({
            "Corr_Z_Y":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # STATUS
    # ============================================================

    BP_ALL_PASS = (
        VALIDATION[
            "BP_Status"
        ]
        .eq("PASS")
        .all()
    )


    VAR_ALL_PASS = (
        VALIDATION[
            "Variance_Status"
        ]
        .eq("PASS")
        .all()
    )


    MORAN_ALL_PASS = (
        VALIDATION[
            "Moran_Status"
        ]
        .eq("PASS")
        .all()
    )


    PATTERN_ALL_PASS = (
        VALIDATION[
            "Pattern_Status"
        ]
        .eq("PASS")
        .all()
    )


    homo = (
        VALIDATION[
            VALIDATION[
                "Error_Structure"
            ]
            ==
            "Homogeneous"
        ]
        .iloc[0]
    )


    hetero = (
        VALIDATION[
            VALIDATION[
                "Error_Structure"
            ]
            ==
            "Heterogeneous"
        ]
        .iloc[0]
    )


    # ============================================================
    # SUMMARY
    # ============================================================

    SUMMARY_GRID = pd.DataFrame({

        "Grid":[
            f"{grid_size}x{grid_size}"
        ],

        "N":[
            N
        ],

        "Rho":[
            rho
        ],

        "Scenario":[
            18
        ],

        "BP_Homo_LM":[
            homo[
                "BP_LM"
            ]
        ],

        "BP_Homo_p":[
            homo[
                "BP_LM_pvalue"
            ]
        ],

        "BP_Homo_Decision":[
            homo[
                "BP_Decision"
            ]
        ],

        "BP_Hetero_LM":[
            hetero[
                "BP_LM"
            ]
        ],

        "BP_Hetero_p":[
            hetero[
                "BP_LM_pvalue"
            ]
        ],

        "BP_Hetero_Decision":[
            hetero[
                "BP_Decision"
            ]
        ],

        "Homo_Var_Ratio":[
            homo[
                "Var_Ratio"
            ]
        ],

        "Hetero_Var_Ratio":[
            hetero[
                "Var_Ratio"
            ]
        ],

        "Mean_Moran_I":[
            VALIDATION[
                "Moran_I_Y"
            ].mean()
        ],

        "Min_Moran_I":[
            VALIDATION[
                "Moran_I_Y"
            ].min()
        ],

        "Max_Moran_I":[
            VALIDATION[
                "Moran_I_Y"
            ].max()
        ],

        "Mean_Corr_Z_Y":[
            VALIDATION[
                "Corr_Z_Y"
            ].mean()
        ],

        "Min_Corr_Z_Y":[
            VALIDATION[
                "Corr_Z_Y"
            ].min()
        ],

        "Max_Corr_Z_Y":[
            VALIDATION[
                "Corr_Z_Y"
            ].max()
        ],

        "BP_All_PASS":[
            BP_ALL_PASS
        ],

        "Variance_All_PASS":[
            VAR_ALL_PASS
        ],

        "Moran_All_PASS":[
            MORAN_ALL_PASS
        ],

        "Pattern_All_PASS":[
            PATTERN_ALL_PASS
        ]

    })


    print("\n")
    print("=" * 150)

    print(
        f"SUMMARY FINAL GRID {grid_size}×{grid_size} — MODERATE"
    )

    print("=" * 150)


    print(
        SUMMARY_GRID

        .round({
            "Rho":2,
            "BP_Homo_LM":4,
            "BP_Homo_p":6,
            "BP_Hetero_LM":4,
            "BP_Hetero_p":6,
            "Homo_Var_Ratio":4,
            "Hetero_Var_Ratio":4,
            "Mean_Moran_I":4,
            "Min_Moran_I":4,
            "Max_Moran_I":4,
            "Mean_Corr_Z_Y":4,
            "Min_Corr_Z_Y":4,
            "Max_Corr_Z_Y":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # SAVE
    # ============================================================

    DATA_FILE = (
        OUTPUT_DIR
        /
        f"DATA_MODERATE_GRID{grid_size}_RHO050.csv"
    )


    DATA.to_csv(
        DATA_FILE,
        index=False
    )


    VALIDATION_FILE = (
        OUTPUT_DIR
        /
        f"VALIDASI_MODERATE_GRID{grid_size}_RHO050.csv"
    )


    VALIDATION.to_csv(
        VALIDATION_FILE,
        index=False
    )


    BP_FILE = (
        OUTPUT_DIR
        /
        f"BREUSCH_PAGAN_GRID{grid_size}.csv"
    )


    BP_CHECK.to_csv(
        BP_FILE,
        index=False
    )


    SUMMARY_FILE = (
        OUTPUT_DIR
        /
        f"SUMMARY_MODERATE_GRID{grid_size}.csv"
    )


    SUMMARY_GRID.to_csv(
        SUMMARY_FILE,
        index=False
    )


    W_INFO.to_csv(
        OUTPUT_DIR
        /
        f"INFO_W_GRID{grid_size}.csv",
        index=False
    )


    for method in W_METHODS:

        sp.save_npz(
            OUTPUT_DIR
            /
            f"W_GRID{grid_size}_{method}.npz",
            W_STORE[
                method
            ]
        )


    print("\n")
    print("=" * 100)

    print(
        f"🎉 GRID {grid_size}×{grid_size} MODERATE SELESAI"
    )

    print("=" * 100)


    print(
        "Breusch-Pagan :",
        "PASS ✅"
        if BP_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Varians Zona  :",
        "PASS ✅"
        if VAR_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Moran's I     :",
        "PASS ✅"
        if MORAN_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Pola Z(s)→Y   :",
        "PASS ✅"
        if PATTERN_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "\n📁 OUTPUT:"
    )

    print(
        OUTPUT_DIR
    )


    gc.collect()


    return {

        "grid":
            grid_size,

        "data":
            DATA,

        "validation":
            VALIDATION,

        "bp":
            BP_CHECK,

        "moran":
            MORAN_CHECK,

        "pattern_check":
            PATTERN_CHECK,

        "summary":
            SUMMARY_GRID,

        "output":
            OUTPUT_DIR

    }


```

```python
# ================================================================
# ================================================================
# STRONG SPATIAL DEPENDENCE
# rho = 0.80
#
# CELL 1
# GRID 10×10
#
# Sekaligus mendefinisikan fungsi untuk Grid10-Grid50
# ================================================================
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import gc
import zipfile

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import matplotlib as mpl

import scipy.sparse as sp

from scipy.spatial import cKDTree
from scipy.sparse.linalg import spsolve
from scipy.interpolate import UnivariateSpline

import statsmodels.api as sm

from statsmodels.stats.diagnostic import (
    het_breuschpagan
)

from pathlib import Path


# ================================================================
# DISPLAY
# ================================================================

pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.width",
    3000
)

pd.set_option(
    "display.max_colwidth",
    None
)


# ================================================================
# 1. KONFIGURASI STRONG
# ================================================================

RHO_STRONG = 0.80

DELTA_SPATIAL = 5.0

LAMBDA_X = 0.80

ERROR_SD = 0.25

SPLINE_SMOOTH = 0.60


PATTERNS = [
    "Hotspot",
    "Koridor",
    "Gelombang"
]


W_METHODS = [
    "KNN5",
    "DISTANCE",
    "IDW"
]


ERROR_TYPES = [
    "Homogeneous",
    "Heterogeneous"
]


MODEL_PREDICTORS = [
    "X1",
    "X2",
    "X3"
]


# ================================================================
# OUTPUT
# ================================================================

BASE_OUTPUT_STRONG = Path(
    "/content/SIMULASI_STRONG_RHO080"
)


BASE_OUTPUT_STRONG.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# STANDARDIZE
# ================================================================

def standardize(v):

    v = np.asarray(
        v,
        dtype=float
    )

    sd = np.std(
        v,
        ddof=0
    )

    if sd <= 1e-12:

        return (
            v
            -
            np.mean(v)
        )

    return (
        (v - np.mean(v))
        /
        sd
    )


# ================================================================
# ROW STANDARDIZE
# ================================================================

def row_standardize(W):

    W = (
        W
        .tocsr()
        .astype(float)
    )

    row_sum = np.asarray(
        W.sum(axis=1)
    ).ravel()

    inv = np.zeros_like(
        row_sum,
        dtype=float
    )

    valid = (
        row_sum > 0
    )

    inv[valid] = (
        1.0
        /
        row_sum[valid]
    )

    return (
        sp.diags(inv)
        @
        W
    ).tocsr()


# ================================================================
# GAUSSIAN HOTSPOT
# ================================================================

def gaussian_hotspot(
    x,
    y,
    cx,
    cy,
    sigma,
    amplitude=1.0
):

    return (
        amplitude
        *
        np.exp(
            -(
                (x-cx)**2
                +
                (y-cy)**2
            )
            /
            (
                2
                *
                sigma**2
            )
        )
    )


# ================================================================
# SMOOTH SPLINE
# ================================================================

def smooth_spline_curve(
    x,
    y,
    smooth_fraction=0.60
):

    temp = pd.DataFrame({
        "x": np.asarray(x),
        "y": np.asarray(y)
    })

    temp["x_round"] = (
        temp["x"]
        .round(8)
    )

    temp = (
        temp

        .groupby(
            "x_round",
            as_index=False
        )

        .agg(
            x=("x", "mean"),
            y=("y", "mean")
        )

        .sort_values("x")
    )

    xs = (
        temp["x"]
        .to_numpy()
    )

    ys = (
        temp["y"]
        .to_numpy()
    )

    smoothing = (
        smooth_fraction
        *
        len(xs)
        *
        max(
            np.var(ys),
            1e-6
        )
    )

    spline = UnivariateSpline(
        xs,
        ys,
        k=3,
        s=smoothing
    )

    xx = np.linspace(
        xs.min(),
        xs.max(),
        400
    )

    yy = spline(xx)

    return xx, yy


# ================================================================
# MORAN'S I
# ================================================================

def moran_I(
    y,
    W
):

    y = np.asarray(
        y,
        dtype=float
    )

    z = (
        y
        -
        np.mean(y)
    )

    denominator = float(
        z @ z
    )

    if denominator <= 0:

        return np.nan

    numerator = float(
        z
        @
        (W @ z)
    )

    S0 = float(
        W.sum()
    )

    return float(
        (len(y)/S0)
        *
        (numerator/denominator)
    )


# ================================================================
# ================================================================
# FUNGSI UTAMA STRONG
# ================================================================
# ================================================================

def run_strong_grid(
    grid_size
):

    rho = RHO_STRONG

    N = (
        grid_size
        *
        grid_size
    )


    OUTPUT_DIR = (
        BASE_OUTPUT_STRONG
        /
        f"GRID{grid_size}"
    )


    OUTPUT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )


    print("\n")
    print("=" * 120)

    print(
        "STRONG SPATIAL DEPENDENCE"
    )

    print(
        f"rho = {rho:.2f}"
    )

    print(
        f"GRID {grid_size} × {grid_size}"
    )

    print(
        f"N = {N}"
    )

    print("=" * 120)


    # ============================================================
    # A. GRID
    # ============================================================

    grid_row = np.repeat(
        np.arange(grid_size),
        grid_size
    )


    grid_col = np.tile(
        np.arange(grid_size),
        grid_size
    )


    x_raw = (
        grid_col
        .astype(float)
    )


    y_raw = (
        grid_row
        .astype(float)
    )


    coords_raw = np.column_stack([
        x_raw,
        y_raw
    ])


    x_norm = (
        x_raw
        /
        (grid_size-1)
    )


    y_norm = (
        y_raw
        /
        (grid_size-1)
    )


    GRID = pd.DataFrame({

        "Location_ID":
            np.arange(
                1,
                N+1
            ),

        "Grid_Row":
            grid_row,

        "Grid_Col":
            grid_col,

        "X_norm":
            x_norm,

        "Y_norm":
            y_norm

    })


    # ============================================================
    # B. HOTSPOT
    # ============================================================

    hotspot_big = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.32,
        cy=0.56,
        sigma=0.17,
        amplitude=1.80
    )


    hotspot_small = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.74,
        cy=0.76,
        sigma=0.085,
        amplitude=1.30
    )


    hotspot_irregular = (
        0.08
        *
        np.sin(
            3*np.pi*x_norm
        )
        *
        np.cos(
            2*np.pi*y_norm
        )
    )


    HOTSPOT_PARTS = {

        "besar":
            standardize(
                hotspot_big
            ),

        "kecil":
            standardize(
                hotspot_small
            ),

        "gabungan":
            standardize(
                hotspot_big
                +
                hotspot_small
                +
                hotspot_irregular
            )
    }


    # ============================================================
    # C. KORIDOR
    # ============================================================

    centerline = (
        0.20

        +

        0.54
        *
        x_norm

        +

        0.10
        *
        np.sin(
            2.6*np.pi*x_norm
            +
            0.40
        )

        -

        0.045
        *
        np.sin(
            5.4*np.pi*x_norm
        )
    )


    corridor_distance = np.abs(
        y_norm
        -
        centerline
    )


    corridor_width = (
        0.065

        +

        0.020
        *
        (
            1
            +
            np.sin(
                3*np.pi*x_norm
            )
        )
    )


    corridor_main = np.exp(
        -(corridor_distance**2)
        /
        (
            2
            *
            corridor_width**2
        )
    )


    taper_left = (
        1
        /
        (
            1
            +
            np.exp(
                -(x_norm-0.10)
                /
                0.025
            )
        )
    )


    taper_right = (
        1
        /
        (
            1
            +
            np.exp(
                (x_norm-0.90)
                /
                0.025
            )
        )
    )


    corridor_main = (
        corridor_main
        *
        taper_left
        *
        taper_right
    )


    corridor_bump1 = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.28,
        cy=0.38,
        sigma=0.085,
        amplitude=0.45
    )


    corridor_bump2 = gaussian_hotspot(
        x_norm,
        y_norm,
        cx=0.72,
        cy=0.67,
        sigma=0.070,
        amplitude=0.35
    )


    corridor_local = (
        corridor_bump1
        +
        corridor_bump2
    )


    corridor_roughness = (
        0.08
        *
        np.sin(
            5*np.pi*x_norm
            +
            1.2*y_norm
        )
    )


    CORRIDOR_PARTS = {

        "utama":
            standardize(
                corridor_main
            ),

        "lokal":
            standardize(
                corridor_local
            ),

        "gabungan":
            standardize(
                1.15*corridor_main
                +
                corridor_local
                +
                corridor_roughness
            )
    }


    # ============================================================
    # D. GELOMBANG
    # ============================================================

    wave_large = (
        1.15
        *
        np.sin(
            2.5*np.pi*x_norm
            +
            0.30
        )

        +

        0.95
        *
        np.cos(
            2.1*np.pi*y_norm
            -
            0.20
        )
    )


    wave_secondary = (
        0.55
        *
        np.sin(
            4.0*np.pi*x_norm
            -
            1.7*np.pi*y_norm
        )
    )


    wave_diagonal = (
        0.40
        *
        np.sin(
            2*np.pi
            *
            (
                x_norm
                +
                y_norm
            )
            +
            0.50
        )
    )


    WAVE_PARTS = {

        "besar":
            standardize(
                wave_large
            ),

        "kecil":
            standardize(
                wave_secondary
                +
                wave_diagonal
            ),

        "gabungan":
            standardize(
                wave_large
                +
                wave_secondary
                +
                wave_diagonal
            )
    }


    Z_PATTERN = {

        "Hotspot":
            HOTSPOT_PARTS[
                "gabungan"
            ],

        "Koridor":
            CORRIDOR_PARTS[
                "gabungan"
            ],

        "Gelombang":
            WAVE_PARTS[
                "gabungan"
            ]

    }


    # ============================================================
    # VISUAL 1 — Z(s)
    # ============================================================

    fig, axes = plt.subplots(
        3,
        3,
        figsize=(12,12)
    )


    VISUAL_COMPONENTS = [

        (
            "Hotspot",
            HOTSPOT_PARTS["besar"],
            HOTSPOT_PARTS["kecil"],
            HOTSPOT_PARTS["gabungan"]
        ),

        (
            "Koridor",
            CORRIDOR_PARTS["utama"],
            CORRIDOR_PARTS["lokal"],
            CORRIDOR_PARTS["gabungan"]
        ),

        (
            "Gelombang",
            WAVE_PARTS["besar"],
            WAVE_PARTS["kecil"],
            WAVE_PARTS["gabungan"]
        )
    ]


    titles = [
        "Struktur Besar/Utama",
        "Struktur Kecil/Lokal",
        "Gabungan Z(s)"
    ]


    for r, (
        pattern,
        p1,
        p2,
        combined
    ) in enumerate(
        VISUAL_COMPONENTS
    ):


        for c, values in enumerate(
            [
                p1,
                p2,
                combined
            ]
        ):


            mat = values.reshape(
                grid_size,
                grid_size
            )


            im = axes[
                r,
                c
            ].imshow(
                mat,
                origin="lower",
                cmap="viridis",
                interpolation="nearest"
            )


            if r == 0:

                axes[
                    r,
                    c
                ].set_title(
                    titles[c],
                    fontweight="bold"
                )


            if c == 0:

                axes[
                    r,
                    c
                ].set_ylabel(
                    pattern,
                    fontweight="bold"
                )


            axes[r,c].set_xticks([])

            axes[r,c].set_yticks([])


            plt.colorbar(
                im,
                ax=axes[r,c],
                fraction=.046,
                pad=.04
            )


    fig.suptitle(
        (
            f"Latent Spatial Pattern Z(s)\n"
            f"Grid {grid_size}×{grid_size} | Strong rho=0.80"
        ),
        fontsize=16,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS1 = (
        OUTPUT_DIR
        /
        "01_LATENT_PATTERN.png"
    )


    plt.savefig(
        FILE_VIS1,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # E. X
    #
    # SEED SAMA DENGAN WEAK DAN MODERATE
    # ============================================================

    rng_x = np.random.default_rng(
        2026
        +
        grid_size
    )


    X_DATA = rng_x.normal(
        0,
        1,
        size=(N,3)
    )


    for j in range(3):

        X_DATA[:,j] = standardize(
            X_DATA[:,j]
        )


    X1 = X_DATA[:,0]

    X2 = X_DATA[:,1]

    X3 = X_DATA[:,2]


    GX_RAW = (
        np.sin(
            1.4*X1
        )

        +

        0.60
        *
        (
            X2**2
            -
            1
        )

        +

        0.60
        *
        np.tanh(
            X3
        )
    )


    GX = standardize(
        GX_RAW
    )


    # ============================================================
    # VISUAL 2 — NONLINEAR
    # ============================================================

    X_DICT = {
        "X1":X1,
        "X2":X2,
        "X3":X3
    }


    fig, axes = plt.subplots(
        1,
        3,
        figsize=(15,4.5)
    )


    for ax, xname in zip(
        axes,
        MODEL_PREDICTORS
    ):


        xx, yy = smooth_spline_curve(
            X_DICT[xname],
            GX,
            smooth_fraction=.25
        )


        ax.scatter(
            X_DICT[xname],
            GX,
            alpha=.30,
            s=18
        )


        ax.plot(
            xx,
            yy,
            linewidth=3
        )


        ax.set_title(
            f"{xname} vs g(X)",
            fontweight="bold"
        )


        ax.set_xlabel(xname)

        ax.set_ylabel("g(X)")

        ax.grid(alpha=.20)


    fig.suptitle(
        (
            "Cek Nonlinearitas Sebenarnya pada DGP\n"
            f"Grid {grid_size}×{grid_size}"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS2 = (
        OUTPUT_DIR
        /
        "02_TRUE_NONLINEAR_SPLINE.png"
    )


    plt.savefig(
        FILE_VIS2,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # F. W
    # ============================================================

    tree = cKDTree(
        coords_raw
    )


    # ---------------- KNN5 ----------------

    _, indices_knn = tree.query(
        coords_raw,
        k=6
    )


    rows_knn = np.repeat(
        np.arange(N),
        5
    )


    cols_knn = (
        indices_knn[:,1:6]
        .reshape(-1)
    )


    vals_knn = np.ones(
        len(rows_knn)
    )


    W_KNN5 = sp.csr_matrix(
        (
            vals_knn,
            (
                rows_knn,
                cols_knn
            )
        ),
        shape=(N,N)
    )


    W_KNN5 = row_standardize(
        W_KNN5
    )


    # ---------------- DISTANCE ----------------

    pairs_distance = np.array(
        list(
            tree.query_pairs(
                r=1.5
            )
        ),
        dtype=int
    )


    if len(pairs_distance) > 0:

        rows_d = np.concatenate([
            pairs_distance[:,0],
            pairs_distance[:,1]
        ])

        cols_d = np.concatenate([
            pairs_distance[:,1],
            pairs_distance[:,0]
        ])

        vals_d = np.ones(
            len(rows_d)
        )

    else:

        rows_d = np.array(
            [],
            dtype=int
        )

        cols_d = np.array(
            [],
            dtype=int
        )

        vals_d = np.array(
            [],
            dtype=float
        )


    W_DISTANCE = sp.csr_matrix(
        (
            vals_d,
            (
                rows_d,
                cols_d
            )
        ),
        shape=(N,N)
    )


    W_DISTANCE = row_standardize(
        W_DISTANCE
    )


    # ---------------- IDW ----------------

    pairs_idw = np.array(
        list(
            tree.query_pairs(
                r=2.5
            )
        ),
        dtype=int
    )


    if len(pairs_idw) > 0:

        i1 = pairs_idw[:,0]

        i2 = pairs_idw[:,1]


        dist_idw = np.linalg.norm(
            coords_raw[i1]
            -
            coords_raw[i2],
            axis=1
        )


        weight_idw = (
            1.0
            /
            dist_idw
        )


        rows_i = np.concatenate([
            i1,
            i2
        ])


        cols_i = np.concatenate([
            i2,
            i1
        ])


        vals_i = np.concatenate([
            weight_idw,
            weight_idw
        ])

    else:

        rows_i = np.array(
            [],
            dtype=int
        )

        cols_i = np.array(
            [],
            dtype=int
        )

        vals_i = np.array(
            [],
            dtype=float
        )


    W_IDW = sp.csr_matrix(
        (
            vals_i,
            (
                rows_i,
                cols_i
            )
        ),
        shape=(N,N)
    )


    W_IDW = row_standardize(
        W_IDW
    )


    W_STORE = {

        "KNN5":
            W_KNN5,

        "DISTANCE":
            W_DISTANCE,

        "IDW":
            W_IDW
    }


    W_INFO_ROWS = []


    for method in W_METHODS:


        counts = np.diff(
            W_STORE[
                method
            ].indptr
        )


        W_INFO_ROWS.append({

            "Grid":
                f"{grid_size}x{grid_size}",

            "W_Method":
                method,

            "Min_Neighbor":
                int(
                    counts.min()
                ),

            "Mean_Neighbor":
                float(
                    counts.mean()
                ),

            "Max_Neighbor":
                int(
                    counts.max()
                )
        })


    W_INFO = pd.DataFrame(
        W_INFO_ROWS
    )


    print("\nINFORMASI W:")

    print(
        W_INFO
        .round(2)
        .to_string(
            index=False
        )
    )


    # ============================================================
    # VISUAL 3 — W
    # ============================================================

    target_index = np.argmin(
        (
            x_norm-.5
        )**2
        +
        (
            y_norm-.5
        )**2
    )


    fig, axes = plt.subplots(
        1,
        3,
        figsize=(15,5)
    )


    for ax, method in zip(
        axes,
        W_METHODS
    ):


        W = W_STORE[
            method
        ].tocsr()


        start = W.indptr[
            target_index
        ]

        end = W.indptr[
            target_index+1
        ]


        neighbors = W.indices[
            start:end
        ]


        ax.scatter(
            x_raw,
            y_raw,
            s=12,
            alpha=.20
        )


        ax.scatter(
            x_raw[neighbors],
            y_raw[neighbors],
            s=70,
            label="Tetangga"
        )


        ax.scatter(
            x_raw[target_index],
            y_raw[target_index],
            marker="*",
            s=220,
            label="Lokasi pusat"
        )


        for j in neighbors:

            ax.plot(
                [
                    x_raw[target_index],
                    x_raw[j]
                ],
                [
                    y_raw[target_index],
                    y_raw[j]
                ],
                linewidth=.8,
                alpha=.50
            )


        ax.set_title(
            (
                f"{method}\n"
                f"k={len(neighbors)}"
            ),
            fontweight="bold"
        )


        ax.set_aspect(
            "equal"
        )


        ax.legend(
            fontsize=7
        )


    fig.suptitle(
        (
            "Struktur Ketetanggaan W\n"
            f"Grid {grid_size}×{grid_size}"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS3 = (
        OUTPUT_DIR
        /
        "03_STRUKTUR_W.png"
    )


    plt.savefig(
        FILE_VIS3,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # G. ERROR
    # ============================================================

    rng_error = np.random.default_rng(
        2027
        +
        grid_size
    )


    BASE_ERROR = rng_error.normal(
        0,
        1,
        N
    )


    def generate_error(
        error_type
    ):


        if (
            error_type
            ==
            "Homogeneous"
        ):

            variance = np.ones(
                N
            )

            error = (
                ERROR_SD
                *
                BASE_ERROR
            )

        else:

            variance = np.where(
                x_norm <= .5,
                .5,
                1.5
            )

            error = (
                ERROR_SD
                *
                np.sqrt(
                    variance
                )
                *
                BASE_ERROR
            )


        return (
            error,
            variance
        )


    # ============================================================
    # H. 18 SKENARIO
    # ============================================================

    ALL_DATA = []


    for pattern in PATTERNS:


        Z_SPATIAL = Z_PATTERN[
            pattern
        ]


        SPATIAL_SIGNAL = (
            DELTA_SPATIAL
            *
            Z_SPATIAL
        )


        for error_type in ERROR_TYPES:


            ERROR, ERROR_VARIANCE = (
                generate_error(
                    error_type
                )
            )


            ETA = (
                SPATIAL_SIGNAL
                +
                LAMBDA_X
                *
                GX
                +
                ERROR
            )


            for w_method in W_METHODS:


                W = W_STORE[
                    w_method
                ]


                A = (
                    sp.eye(
                        N,
                        format="csr"
                    )
                    -
                    rho
                    *
                    W
                )


                Y = spsolve(
                    A,
                    ETA
                )


                Y = np.asarray(
                    Y,
                    dtype=float
                )


                moran_y = moran_I(
                    Y,
                    W
                )


                corr_zy = float(
                    np.corrcoef(
                        Z_SPATIAL,
                        Y
                    )[0,1]
                )


                scenario_id = (
                    f"G{grid_size}_"
                    f"{pattern.upper()}_"
                    f"{w_method}_"
                    f"{'HOMO' if error_type == 'Homogeneous' else 'HETERO'}_"
                    f"RHO80"
                )


                temp = GRID.copy()


                temp[
                    "Scenario_ID"
                ] = scenario_id


                temp[
                    "Grid"
                ] = (
                    f"{grid_size}x{grid_size}"
                )


                temp[
                    "N"
                ] = N


                temp[
                    "Strength"
                ] = "Strong"


                temp[
                    "Rho"
                ] = rho


                temp[
                    "Pattern"
                ] = pattern


                temp[
                    "W_Method"
                ] = w_method


                temp[
                    "Error_Structure"
                ] = error_type


                temp[
                    "Z_spatial_latent"
                ] = Z_SPATIAL


                temp["X1"] = X1

                temp["X2"] = X2

                temp["X3"] = X3


                temp[
                    "gX_nonlinear"
                ] = GX


                temp[
                    "Spatial_Signal"
                ] = SPATIAL_SIGNAL


                temp[
                    "Error"
                ] = ERROR


                temp[
                    "Error_Variance"
                ] = ERROR_VARIANCE


                temp[
                    "Eta"
                ] = ETA


                temp[
                    "Y"
                ] = Y


                temp[
                    "Moran_I_Y"
                ] = moran_y


                temp[
                    "Corr_Z_Y"
                ] = corr_zy


                ALL_DATA.append(
                    temp
                )


    DATA = pd.concat(
        ALL_DATA,
        ignore_index=True
    )


    assert (
        DATA[
            "Scenario_ID"
        ].nunique()
        ==
        18
    )


    # ============================================================
    # VISUAL 4 — 18 Y
    # ============================================================

    rows_plot = [
        (
            p,
            e
        )
        for p in PATTERNS
        for e in ERROR_TYPES
    ]


    norm = mpl.colors.Normalize(
        vmin=float(
            DATA["Y"].min()
        ),
        vmax=float(
            DATA["Y"].max()
        )
    )


    fig, axes = plt.subplots(
        6,
        3,
        figsize=(14,24)
    )


    last_im = None


    for r, (
        pattern,
        error_type
    ) in enumerate(
        rows_plot
    ):


        for c, w_method in enumerate(
            W_METHODS
        ):


            d = (
                DATA[
                    (
                        DATA[
                            "Pattern"
                        ]
                        ==
                        pattern
                    )

                    &

                    (
                        DATA[
                            "Error_Structure"
                        ]
                        ==
                        error_type
                    )

                    &

                    (
                        DATA[
                            "W_Method"
                        ]
                        ==
                        w_method
                    )
                ]

                .sort_values(
                    "Location_ID"
                )
            )


            YMAT = (
                d["Y"]
                .to_numpy()
                .reshape(
                    grid_size,
                    grid_size
                )
            )


            ax = axes[
                r,
                c
            ]


            last_im = ax.imshow(
                YMAT,
                origin="lower",
                cmap="viridis",
                norm=norm,
                interpolation="nearest"
            )


            if r == 0:

                ax.set_title(
                    w_method,
                    fontweight="bold"
                )


            if c == 0:

                err = (
                    "Homo"
                    if error_type
                    ==
                    "Homogeneous"
                    else
                    "Hetero"
                )


                ax.set_ylabel(
                    (
                        f"{pattern}\n"
                        f"{err}"
                    ),
                    fontweight="bold"
                )


            if (
                error_type
                ==
                "Heterogeneous"
            ):

                ax.axvline(
                    grid_size/2-.5,
                    linestyle="--",
                    linewidth=1.5
                )


            counts = np.diff(
                W_STORE[
                    w_method
                ].indptr
            )


            ax.text(
                .5,
                -.08,
                (
                    f"ρ=.80 | "
                    f"k̄={counts.mean():.2f}"
                    "\n"
                    f"Moran I={d['Moran_I_Y'].iloc[0]:.3f}"
                    f" | Corr(Z,Y)={d['Corr_Z_Y'].iloc[0]:.3f}"
                ),
                transform=ax.transAxes,
                ha="center",
                fontsize=7.5
            )


            ax.set_xticks([])

            ax.set_yticks([])


    fig.colorbar(
        last_im,
        ax=axes,
        fraction=.015,
        pad=.02,
        label="Nilai Y"
    )


    fig.suptitle(
        (
            f"18 Skenario Y — Grid {grid_size}×{grid_size}\n"
            "STRONG Spatial Dependence | rho=0.80"
        ),
        fontsize=17,
        fontweight="bold",
        y=.995
    )


    FILE_VIS4 = (
        OUTPUT_DIR
        /
        "04_Y_18_SKENARIO.png"
    )


    plt.savefig(
        FILE_VIS4,
        dpi=180,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # VISUAL 5 — Z VS Y
    # ============================================================

    fig, axes = plt.subplots(
        3,
        2,
        figsize=(8,12)
    )


    for r, pattern in enumerate(
        PATTERNS
    ):


        d = (
            DATA[
                (
                    DATA[
                        "Pattern"
                    ]
                    ==
                    pattern
                )

                &

                (
                    DATA[
                        "W_Method"
                    ]
                    ==
                    "KNN5"
                )

                &

                (
                    DATA[
                        "Error_Structure"
                    ]
                    ==
                    "Homogeneous"
                )
            ]

            .sort_values(
                "Location_ID"
            )
        )


        ZMAT = (
            d[
                "Z_spatial_latent"
            ]
            .to_numpy()
            .reshape(
                grid_size,
                grid_size
            )
        )


        YMAT = (
            d["Y"]
            .to_numpy()
            .reshape(
                grid_size,
                grid_size
            )
        )


        im1 = axes[
            r,
            0
        ].imshow(
            ZMAT,
            origin="lower",
            cmap="viridis"
        )


        axes[
            r,
            0
        ].set_title(
            (
                f"{pattern}\n"
                "Z(s)"
            ),
            fontweight="bold"
        )


        plt.colorbar(
            im1,
            ax=axes[r,0],
            fraction=.046,
            pad=.04
        )


        im2 = axes[
            r,
            1
        ].imshow(
            YMAT,
            origin="lower",
            cmap="viridis"
        )


        axes[
            r,
            1
        ].set_title(
            (
                f"{pattern}\n"
                f"Y | Corr="
                f"{d['Corr_Z_Y'].iloc[0]:.3f}"
            ),
            fontweight="bold"
        )


        plt.colorbar(
            im2,
            ax=axes[r,1],
            fraction=.046,
            pad=.04
        )


        for c in range(2):

            axes[
                r,
                c
            ].set_xticks([])

            axes[
                r,
                c
            ].set_yticks([])


    fig.suptitle(
        (
            "Kesesuaian Pola Z(s) → Y\n"
            f"Grid {grid_size}×{grid_size} | Strong"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS5 = (
        OUTPUT_DIR
        /
        "05_Z_VS_Y.png"
    )


    plt.savefig(
        FILE_VIS5,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # VISUAL 6 — SPLINE
    # ============================================================

    fig, axes = plt.subplots(
        3,
        3,
        figsize=(15,13)
    )


    for r, pattern in enumerate(
        PATTERNS
    ):


        d = (
            DATA[
                (
                    DATA[
                        "Pattern"
                    ]
                    ==
                    pattern
                )

                &

                (
                    DATA[
                        "W_Method"
                    ]
                    ==
                    "KNN5"
                )

                &

                (
                    DATA[
                        "Error_Structure"
                    ]
                    ==
                    "Homogeneous"
                )
            ]

            .sort_values(
                "Location_ID"
            )
        )


        for c, xname in enumerate(
            MODEL_PREDICTORS
        ):


            xv = (
                d[xname]
                .to_numpy()
            )


            yv = (
                d["Y"]
                .to_numpy()
            )


            xx, yy = smooth_spline_curve(
                xv,
                yv,
                smooth_fraction=
                    SPLINE_SMOOTH
            )


            axes[
                r,
                c
            ].scatter(
                xv,
                yv,
                alpha=.25,
                s=18
            )


            axes[
                r,
                c
            ].plot(
                xx,
                yy,
                linewidth=3
            )


            if r == 0:

                axes[
                    r,
                    c
                ].set_title(
                    xname,
                    fontweight="bold"
                )


            if c == 0:

                axes[
                    r,
                    c
                ].set_ylabel(
                    (
                        f"{pattern}\n"
                        "Y"
                    ),
                    fontweight="bold"
                )


            axes[
                r,
                c
            ].set_xlabel(
                xname
            )


            axes[
                r,
                c
            ].grid(
                alpha=.20
            )


    fig.suptitle(
        (
            "Smooth Spline Prediktor terhadap Y\n"
            f"Grid {grid_size}×{grid_size} | KNN5 | Homo | rho=.80"
        ),
        fontsize=15,
        fontweight="bold"
    )


    plt.tight_layout()


    FILE_VIS6 = (
        OUTPUT_DIR
        /
        "06_SMOOTH_SPLINE_X_Y.png"
    )


    plt.savefig(
        FILE_VIS6,
        dpi=200,
        bbox_inches="tight"
    )


    plt.show()

    plt.close(fig)


    # ============================================================
    # I. VALIDASI BP + MORAN + PATTERN
    # ============================================================

    VALIDATION_ROWS = []


    for scenario_id in (
        DATA[
            "Scenario_ID"
        ].drop_duplicates()
    ):


        d = (
            DATA[
                DATA[
                    "Scenario_ID"
                ]
                ==
                scenario_id
            ]

            .sort_values(
                "Location_ID"
            )

            .reset_index(
                drop=True
            )
        )


        true_mean = (
            d[
                "Spatial_Signal"
            ].to_numpy()

            +

            LAMBDA_X
            *
            d[
                "gX_nonlinear"
            ].to_numpy()
        )


        residual = (
            d[
                "Eta"
            ].to_numpy()
            -
            true_mean
        )


        zone2 = (
            d[
                "X_norm"
            ].to_numpy()
            >
            .5
        ).astype(int)


        X_BP = sm.add_constant(
            zone2
        )


        (
            BP_LM,
            BP_LM_p,
            BP_F,
            BP_F_p
        ) = het_breuschpagan(
            residual,
            X_BP
        )


        error_type = (
            d[
                "Error_Structure"
            ].iloc[0]
        )


        if (
            error_type
            ==
            "Homogeneous"
        ):

            BP_EXPECTED = (
                "Homoskedastik"
            )

            BP_STATUS = (
                "PASS"
                if (
                    BP_LM_p > .05
                    and
                    BP_F_p > .05
                )
                else
                "CHECK"
            )

        else:

            BP_EXPECTED = (
                "Heteroskedastik"
            )

            BP_STATUS = (
                "PASS"
                if (
                    BP_LM_p < .05
                    and
                    BP_F_p < .05
                )
                else
                "CHECK"
            )


        BP_DECISION = (
            "Heteroskedastik"
            if BP_LM_p < .05
            else
            "Homoskedastik"
        )


        var1 = np.var(
            residual[
                zone2 == 0
            ],
            ddof=1
        )


        var2 = np.var(
            residual[
                zone2 == 1
            ],
            ddof=1
        )


        var_ratio = (
            var2
            /
            var1
        )


        if (
            error_type
            ==
            "Homogeneous"
        ):

            VAR_STATUS = (
                "PASS"
                if (
                    .50
                    <=
                    var_ratio
                    <=
                    2.00
                )
                else
                "CHECK"
            )

        else:

            VAR_STATUS = (
                "PASS"
                if var_ratio > 1.50
                else
                "CHECK"
            )


        moran_value = float(
            d[
                "Moran_I_Y"
            ].iloc[0]
        )


        corr_value = float(
            d[
                "Corr_Z_Y"
            ].iloc[0]
        )


        VALIDATION_ROWS.append({

            "Pattern":
                d[
                    "Pattern"
                ].iloc[0],

            "W_Method":
                d[
                    "W_Method"
                ].iloc[0],

            "Error_Structure":
                error_type,

            "Rho":
                rho,

            "BP_LM":
                BP_LM,

            "BP_LM_pvalue":
                BP_LM_p,

            "BP_F":
                BP_F,

            "BP_F_pvalue":
                BP_F_p,

            "BP_Expected":
                BP_EXPECTED,

            "BP_Decision":
                BP_DECISION,

            "BP_Status":
                BP_STATUS,

            "Var_Zone1":
                var1,

            "Var_Zone2":
                var2,

            "Var_Ratio":
                var_ratio,

            "Variance_Status":
                VAR_STATUS,

            "Moran_I_Y":
                moran_value,

            "Moran_Status":
                (
                    "PASS"
                    if moran_value > 0
                    else
                    "CHECK"
                ),

            "Corr_Z_Y":
                corr_value,

            "Pattern_Status":
                (
                    "PASS"
                    if corr_value >= .75
                    else
                    "CHECK"
                )

        })


    VALIDATION = pd.DataFrame(
        VALIDATION_ROWS
    )


    # ============================================================
    # BREUSCH PAGAN TABLE
    # ============================================================

    BP_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "BP_LM",
                "BP_LM_pvalue",
                "BP_F",
                "BP_F_pvalue",
                "BP_Expected",
                "BP_Decision",
                "BP_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print("\n")
    print("=" * 130)

    print(
        f"CEK BREUSCH-PAGAN — GRID {grid_size}×{grid_size} — STRONG rho=0.80:"
    )

    print("=" * 130)


    print(
        BP_CHECK

        .round({
            "BP_LM":4,
            "BP_LM_pvalue":6,
            "BP_F":4,
            "BP_F_pvalue":6
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # VARIANCE
    # ============================================================

    VAR_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Var_Zone1",
                "Var_Zone2",
                "Var_Ratio",
                "Variance_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print(
        "\nCEK VARIANS ZONA:"
    )


    print(
        VAR_CHECK

        .round({
            "Var_Zone1":6,
            "Var_Zone2":6,
            "Var_Ratio":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # MORAN
    # ============================================================

    MORAN_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Rho",
                "Moran_I_Y",
                "Moran_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print("\n")
    print("=" * 100)

    print(
        f"CEK MORAN — GRID {grid_size}×{grid_size} — STRONG:"
    )

    print("=" * 100)


    print(
        MORAN_CHECK

        .round({
            "Rho":2,
            "Moran_I_Y":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # Z(s) -> Y
    # ============================================================

    PATTERN_CHECK = (
        VALIDATION[
            [
                "Pattern",
                "W_Method",
                "Error_Structure",
                "Corr_Z_Y",
                "Pattern_Status"
            ]
        ]

        .sort_values(
            [
                "Pattern",
                "W_Method",
                "Error_Structure"
            ]
        )

        .reset_index(
            drop=True
        )
    )


    print("\n")
    print("=" * 100)

    print(
        f"KESESUAIAN POLA Z(s) → Y — GRID {grid_size}×{grid_size}:"
    )

    print("=" * 100)


    print(
        PATTERN_CHECK

        .round({
            "Corr_Z_Y":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # FINAL STATUS
    # ============================================================

    BP_ALL_PASS = (
        VALIDATION[
            "BP_Status"
        ]
        .eq("PASS")
        .all()
    )


    VAR_ALL_PASS = (
        VALIDATION[
            "Variance_Status"
        ]
        .eq("PASS")
        .all()
    )


    MORAN_ALL_PASS = (
        VALIDATION[
            "Moran_Status"
        ]
        .eq("PASS")
        .all()
    )


    PATTERN_ALL_PASS = (
        VALIDATION[
            "Pattern_Status"
        ]
        .eq("PASS")
        .all()
    )


    homo = (
        VALIDATION[
            VALIDATION[
                "Error_Structure"
            ]
            ==
            "Homogeneous"
        ]
        .iloc[0]
    )


    hetero = (
        VALIDATION[
            VALIDATION[
                "Error_Structure"
            ]
            ==
            "Heterogeneous"
        ]
        .iloc[0]
    )


    SUMMARY_GRID = pd.DataFrame({

        "Grid":[
            f"{grid_size}x{grid_size}"
        ],

        "N":[N],

        "Rho":[rho],

        "Scenario":[18],

        "BP_Homo_LM":[
            homo[
                "BP_LM"
            ]
        ],

        "BP_Homo_p":[
            homo[
                "BP_LM_pvalue"
            ]
        ],

        "BP_Homo_Decision":[
            homo[
                "BP_Decision"
            ]
        ],

        "BP_Hetero_LM":[
            hetero[
                "BP_LM"
            ]
        ],

        "BP_Hetero_p":[
            hetero[
                "BP_LM_pvalue"
            ]
        ],

        "BP_Hetero_Decision":[
            hetero[
                "BP_Decision"
            ]
        ],

        "Homo_Var_Ratio":[
            homo[
                "Var_Ratio"
            ]
        ],

        "Hetero_Var_Ratio":[
            hetero[
                "Var_Ratio"
            ]
        ],

        "Mean_Moran_I":[
            VALIDATION[
                "Moran_I_Y"
            ].mean()
        ],

        "Min_Moran_I":[
            VALIDATION[
                "Moran_I_Y"
            ].min()
        ],

        "Max_Moran_I":[
            VALIDATION[
                "Moran_I_Y"
            ].max()
        ],

        "Mean_Corr_Z_Y":[
            VALIDATION[
                "Corr_Z_Y"
            ].mean()
        ],

        "Min_Corr_Z_Y":[
            VALIDATION[
                "Corr_Z_Y"
            ].min()
        ],

        "Max_Corr_Z_Y":[
            VALIDATION[
                "Corr_Z_Y"
            ].max()
        ],

        "BP_All_PASS":[
            BP_ALL_PASS
        ],

        "Variance_All_PASS":[
            VAR_ALL_PASS
        ],

        "Moran_All_PASS":[
            MORAN_ALL_PASS
        ],

        "Pattern_All_PASS":[
            PATTERN_ALL_PASS
        ]

    })


    print("\n")
    print("=" * 150)

    print(
        f"SUMMARY FINAL GRID {grid_size}×{grid_size} — STRONG"
    )

    print("=" * 150)


    print(
        SUMMARY_GRID

        .round({
            "Rho":2,
            "BP_Homo_LM":4,
            "BP_Homo_p":6,
            "BP_Hetero_LM":4,
            "BP_Hetero_p":6,
            "Homo_Var_Ratio":4,
            "Hetero_Var_Ratio":4,
            "Mean_Moran_I":4,
            "Min_Moran_I":4,
            "Max_Moran_I":4,
            "Mean_Corr_Z_Y":4,
            "Min_Corr_Z_Y":4,
            "Max_Corr_Z_Y":4
        })

        .to_string(
            index=False
        )
    )


    # ============================================================
    # SAVE
    # ============================================================

    DATA_FILE = (
        OUTPUT_DIR
        /
        f"DATA_STRONG_GRID{grid_size}_RHO080.csv"
    )


    DATA.to_csv(
        DATA_FILE,
        index=False
    )


    VALIDATION.to_csv(
        OUTPUT_DIR
        /
        f"VALIDASI_STRONG_GRID{grid_size}_RHO080.csv",
        index=False
    )


    BP_CHECK.to_csv(
        OUTPUT_DIR
        /
        f"BREUSCH_PAGAN_GRID{grid_size}.csv",
        index=False
    )


    SUMMARY_GRID.to_csv(
        OUTPUT_DIR
        /
        f"SUMMARY_STRONG_GRID{grid_size}.csv",
        index=False
    )


    W_INFO.to_csv(
        OUTPUT_DIR
        /
        f"INFO_W_GRID{grid_size}.csv",
        index=False
    )


    for method in W_METHODS:

        sp.save_npz(
            OUTPUT_DIR
            /
            f"W_GRID{grid_size}_{method}.npz",
            W_STORE[
                method
            ]
        )


    print("\n")
    print("=" * 100)

    print(
        f"🎉 GRID {grid_size}×{grid_size} STRONG SELESAI"
    )

    print("=" * 100)


    print(
        "Breusch-Pagan :",
        "PASS ✅"
        if BP_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Varians Zona  :",
        "PASS ✅"
        if VAR_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Moran's I     :",
        "PASS ✅"
        if MORAN_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "Pola Z(s)→Y   :",
        "PASS ✅"
        if PATTERN_ALL_PASS
        else
        "CHECK ⚠️"
    )


    print(
        "\n📁 OUTPUT:"
    )

    print(
        OUTPUT_DIR
    )


    gc.collect()


    return {

        "grid":
            grid_size,

        "data":
            DATA,

        "validation":
            VALIDATION,

        "bp":
            BP_CHECK,

        "moran":
            MORAN_CHECK,

        "pattern_check":
            PATTERN_CHECK,

        "summary":
            SUMMARY_GRID,

        "output":
            OUTPUT_DIR

    }


```
